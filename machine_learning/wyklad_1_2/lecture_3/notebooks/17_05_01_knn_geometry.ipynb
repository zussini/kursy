```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ml_teaching_pack import make_dataset, list_datasets, curriculum_table
from ml_teaching_pack import split_scale, pca_project, add_radial_feature
from ml_teaching_pack import get_model, model_library
from ml_teaching_pack import compare_models, run_basic_experiment
from ml_teaching_pack.visualization import plot_2d_points, plot_decision_boundary_2d, plot_pca_variance, plot_pca_scatter
```

```python
ds = make_dataset('moons', n_samples=500, random_state=42)
X, y = ds.as_xy()
plot_2d_points(X, y, title=ds.description)
plt.show()
```

## Wpływ liczby sąsiadów `k`

```python
from sklearn.neighbors import KNeighborsClassifier

split = split_scale(X, y, scale=True, random_state=42)
for k in [1, 7, 31]:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(split.X_train, split.y_train)
    ax = plot_decision_boundary_2d(model, split.X_train, split.y_train, title=f'KNN, k={k}')
    plt.show()
```
