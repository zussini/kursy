# 06 — Manifold learning

Pytanie: **co jeśli dane leżą na zakrzywionej niskowymiarowej powierzchni?**

PCA zakłada liniową podprzestrzeń. Manifold learning zakłada, że struktura może być zakrzywiona, ale lokalnie sensowna.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ml_teaching_pack import make_dataset, list_datasets, curriculum_table
from ml_teaching_pack import split_scale, pca_project, add_radial_feature
from ml_teaching_pack import get_model, model_library
from ml_teaching_pack import compare_models, run_basic_experiment
from ml_teaching_pack.visualization import plot_2d_points, plot_decision_boundary_2d, plot_pca_variance, plot_pca_scatter
from ml_teaching_pack.manifold import embed
```

```python
ds = make_dataset('swiss_roll', n_samples=900, random_state=42)
X, y = ds.as_xy()
print(ds.description)
```

```python
for method in ['pca', 'isomap', 'lle']:
    Z, model = embed(X, method=method, n_components=2, random_state=42)
    plot_2d_points(Z, ds.metadata['t'], title=f'{method}: Swiss roll embedding')
    plt.show()
```

## t-SNE / UMAP

t-SNE i UMAP są świetne wizualizacyjnie, ale trzeba ostrożnie interpretować odległości globalne. UMAP jest opcjonalny.

```python
Z, model = embed(X, method='tsne', n_components=2, random_state=42, perplexity=35)
plot_2d_points(Z, ds.metadata['t'], title='t-SNE: Swiss roll embedding')
plt.show()

try:
    Z, model = embed(X, method='umap_optional', n_components=2, random_state=42)
    plot_2d_points(Z, ds.metadata['t'], title='UMAP optional: Swiss roll embedding')
    plt.show()
except ImportError as e:
    print(e)
```
