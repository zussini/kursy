# 07 — Full comparison pipeline

Ten notebook pokazuje jeden uniwersalny schemat eksperymentu na wielu datasetach.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ml_teaching_pack import make_dataset, list_datasets, curriculum_table
from ml_teaching_pack import split_scale, pca_project, add_radial_feature
from ml_teaching_pack import get_model, model_library
from ml_teaching_pack import compare_models, run_basic_experiment
from ml_teaching_pack.visualization import plot_2d_points, plot_decision_boundary_2d, plot_pca_variance, plot_pca_scatter
```

```python
all_rows = []
for dataset_name in ['blobs_margin', 'moons', 'circles', 'iris', 'high_dim_redundant']:
    ds = make_dataset(dataset_name, n_samples=600, random_state=42)
    split = split_scale(ds.X, ds.y, scale=True, random_state=42)
    result = compare_models(model_library('full'), split.X_train, split.X_test, split.y_train, split.y_test)
    result['dataset'] = dataset_name
    all_rows.append(result)

summary = pd.concat(all_rows, ignore_index=True)
summary.sort_values(['dataset', 'balanced_accuracy'], ascending=[True, False])
```

## Prosty ranking: najlepszy model per dataset
