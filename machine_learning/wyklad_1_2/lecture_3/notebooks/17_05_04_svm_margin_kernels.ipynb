```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from ml_teaching_pack import make_dataset, list_datasets, curriculum_table
from ml_teaching_pack import split_scale, pca_project, add_radial_feature
from ml_teaching_pack import get_model, model_library
from ml_teaching_pack import compare_models, run_basic_experiment
from ml_teaching_pack.visualization import plot_2d_points, plot_decision_boundary_2d, plot_pca_variance, plot_pca_scatter
```

## Linear SVM na danych prawie liniowo separowalnych

```python
ds = make_dataset('blobs_margin', n_samples=500, random_state=42)
X, y = ds.as_xy()
split = split_scale(X, y, scale=True, random_state=42)

model = get_model('linear_svm', C=1.0)
model.fit(split.X_train, split.y_train)
plot_decision_boundary_2d(model, split.X_train, split.y_train, title='Linear SVM on blobs')
plt.show()
```

## Linear SVM vs RBF SVM na `circles`

```python
ds = make_dataset('circles', n_samples=500, random_state=42)
X, y = ds.as_xy()
split = split_scale(X, y, scale=True, random_state=42)

for name in ['linear_svm', 'svm_rbf']:
    model = get_model(name, random_state=42)
    model.fit(split.X_train, split.y_train)
    plot_decision_boundary_2d(model, split.X_train, split.y_train, title=name)
    plt.show()
```

## Jawna transformacja cech jako wstęp do kerneli

Dla okręgów cecha `r² = x1² + x2²` bardzo pomaga. Kernel robi podobny ruch niejawnie: pracuje w bogatszej przestrzeni cech bez ręcznego wypisywania wszystkich transformacji.

```python
X_r = add_radial_feature(X)
split_r = split_scale(X_r, y, scale=True, random_state=42)
models = {'logreg_with_r2': get_model('logreg'), 'linear_svm_with_r2': get_model('linear_svm')}
compare_models(models, split_r.X_train, split_r.X_test, split_r.y_train, split_r.y_test)
```
