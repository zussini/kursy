```python
import numpy as np
import matplotlib.pyplot as plt
```

## Zadanie 1. Line plot
Narysuj `sin(x)` dla `x` od 0 do `2*pi`.


```python
x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

plt.figure(figsize=(6,3))
plt.plot(x, y)
plt.title("sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

## Zadanie 2. Histogram
Narysuj histogram kilku przykładowych czasów końcowych.


```python
times = np.array([7600, 7300, 8100, 7900, 7450, 7200, 7000, 8350])

plt.figure(figsize=(6,3))
plt.hist(times, bins=5)
plt.title("Rozkład czasów końcowych")
plt.xlabel("sekundy")
plt.ylabel("liczba zawodników")
plt.show()
```

## Zadanie 3. Scatter
Narysuj zależność: wiek minimalny grupy vs średni czas.


```python
age_min = np.array([25, 30, 35, 40])
mean_time = np.array([7450, 7325, 7500, 7600])

plt.figure(figsize=(6,3))
plt.scatter(age_min, mean_time)
plt.title("Wiek a średni czas")
plt.xlabel("dolna granica wieku")
plt.ylabel("średni czas [s]")
plt.show()
```

## Zadanie 4. Bar plot
Narysuj słupki dla średnich ocen domen ankiety.


```python
domains = ["Python", "Data", "SQL", "Math", "ML"]
means = [3.93, 2.94, 4.39, 3.93, 3.27]

plt.figure(figsize=(7,3.5))
plt.bar(domains, means)
plt.ylim(0,5)
plt.title("Średnie domen ankiety")
plt.ylabel("średnia samoocena")
plt.show()
```
