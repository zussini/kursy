# DEMO pliku (zanim przejdziemy do pythona tutaj, zrobić pierwszą część w goegebrze):

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def pokaż_funkcję_i_całkę(f, x_min, x_max, a, b, title):
    x = np.linspace(x_min, x_max, 800)
    y = np.array([f(t) for t in x])

    integral_value, _ = quad(f, a, b)

    plt.figure(figsize=(7, 4))
    plt.plot(x, y, label=title)
    plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
    plt.axhline(0)
    plt.axvline(a, linestyle='--')
    plt.axvline(b, linestyle='--')
    plt.title(f'{title}, całka = {integral_value:.4f}')
    plt.legend()
    plt.show()

    print("Całka oznaczona:", integral_value)

def gauss(x):
    return 1 / np.sqrt(2 * np.pi) * np.exp(-x**2 / 2)

pokaż_funkcję_i_całkę(gauss, -4, 4, -10, 3, 'Gauss standardowy')
````

* **część A**: co warto pokazać w GeoGebrze i jak to krótko objaśnić,
* **część B**: prosty kod w Pythonie, krok po kroku, do tych samych idei.

## Część A. Rozkłady inne niż Gauss, które warto pokazać w GeoGebrze

---

## 1. Rozkład jednostajny

To najprostszy rozkład ciągły.

### Postać

Na przedziale $[a,b]$:

$$
f(x)=\frac{1}{b-a}
$$

dla $x\in[a,b]$, a poza tym przedziałem:

$$
f(x)=0.
$$

### Intuicja

Każda wartość z przedziału jest tak samo prawdopodobna.
Pole całego prostokąta musi wynosić 1.

### Co pokazuje dydaktycznie

To bardzo dobry pierwszy kontrast wobec Gaussa, bo:

* nie ma dzwonowatego kształtu,
* gęstość jest stała,
* prawdopodobieństwo nadal liczymy jako pole pod wykresem.

### W GeoGebrze

Na przykład dla $[0,4]$:

```text
f(x)=If(0<=x<=4,1/4,0)
Integral(f,0,4)
Integral(f,1,3)
```

### Krótka notka

To rozkład, w którym wszystkie wartości z pewnego przedziału są jednakowo „uprzywilejowane”. Całka na całym przedziale daje 1, a całka na podprzedziale daje prawdopodobieństwo trafienia do tego fragmentu.

---

## 2. Rozkład wykładniczy

Bardzo dobry do pokazania asymetrii.

### Postać

Dla $\lambda>0$:

$$
f(x)=\lambda e^{-\lambda x}, \qquad x\ge 0
$$

oraz $f(x)=0$ dla $x<0$.

### Intuicja

Duże wartości są coraz rzadsze.
To rozkład malejący, niesymetryczny.

### Co pokazuje dydaktycznie

* nie każdy rozkład jest symetryczny,
* pole nadal daje prawdopodobieństwo,
* można mówić o „ogonach” rozkładu.

### W GeoGebrze

Dla $\lambda=1$:

```text
f(x)=If(x>=0,exp(-x),0)
Integral(f,0,10)
Integral(f,0,1)
Integral(f,1,3)
```

Dla wersji ogólnej:

```text
lambda=1
f(x)=If(x>=0,lambda*exp(-lambda*x),0)
```

### Krótka notka

To rozkład ciągły opisujący sytuacje, w których małe wartości są częste, a duże coraz rzadsze. Pole pod wykresem między dwiema liczbami nadal oznacza prawdopodobieństwo.

---

## 3. Rozkład Laplace’a

Bardzo ciekawy jako porównanie z Gaussem.

### Postać

Dla środka $\mu$ i skali $b>0$:

$$
f(x)=\frac{1}{2b}\exp\left(-\frac{|x-\mu|}{b}\right)
$$

### Intuicja

Jest symetryczny jak Gauss, ale:

* ma ostrzejszy szczyt,
* ma cięższe ogony.

### Co pokazuje dydaktycznie

* dwa rozkłady mogą być symetryczne, ale wyglądać inaczej,
* ogony rozkładu mają znaczenie,
* nie każda „górka” jest Gaussem.

### W GeoGebrze

Dla $\mu=0$, $b=1$:

```text
f(x)=1/2*exp(-abs(x))
Integral(f,-10,10)
Integral(f,-1,1)
```

### Krótka notka

To rozkład symetryczny podobny do Gaussa, ale bardziej „spiczasty” w środku i z wolniej zanikającymi ogonami. Dobrze pokazuje, że rozkłady o podobnym środku mogą różnić się częstością skrajnych wartości.

---

## 4. Rozkład Cauchy’ego

Bardzo dobry jako ostrzeżenie, że nie każdy rozkład zachowuje się „grzecznie”.

### Postać

Dla standardowego przypadku:

$$
f(x)=\frac{1}{\pi(1+x^2)}
$$

### Intuicja

Jest symetryczny, ale ma bardzo ciężkie ogony.

### Co pokazuje dydaktycznie

* nie każdy rozkład ma tak wygodne własności jak Gauss,
* ogony mogą być bardzo ważne,
* można porównać wizualnie z Gaussem.

### W GeoGebrze

```text
f(x)=1/(pi*(1+x^2))
Integral(f,-20,20)
Integral(f,-1,1)
```

### Krótka notka

To rozkład symetryczny, ale z bardzo ciężkimi ogonami. W praktyce pokazuje, że intuicja „większość wartości jest blisko środka” nie zawsze działa tak dobrze jak dla rozkładu normalnego.

---

## 5. Rozkład trójkątny

Świetny pomost między geometrią a statystyką.

### Postać

Na przykład symetryczny na $[0,2]$ z wierzchołkiem w 1:

$$
f(x)=
\begin{cases}
x, & 0\le x\le 1 \
2-x, & 1 < x \le 2 \
0, & \text{poza tym}
\end{cases}
$$

### Intuicja

Najbardziej prawdopodobne są wartości blisko środka, ale wszystko ma bardzo prostą, odcinkową postać.

### Co pokazuje dydaktycznie

* można mieć rozkład gęstości zbudowany z prostych fragmentów,
* pole nadal wynosi 1,
* łatwo liczyć całki ręcznie i geometrycznie.

### W GeoGebrze

```text
f(x)=If(0<=x<=1,x,If(1<x<=2,2-x,0))
Integral(f,0,2)
Integral(f,0.5,1.5)
```

### Krótka notka

To prosty rozkład ciągły złożony z dwóch odcinków prostych. Jest bardzo dobry dydaktycznie, bo można jednocześnie liczyć pola geometrycznie i całkowo.

---

## Jaką kolejność rozkładów bym wybrał

Po Gaussie dałbym:

1. **jednostajny** jako najprostszy,
2. **trójkątny** jako prosty, ale już nie stały,
3. **wykładniczy** jako asymetryczny,
4. **Laplace** jako porównanie do Gaussa,
5. ewentualnie **Cauchy** jako ciekawostkę o ciężkich ogonach.

To daje bardzo naturalny rozwój intuicji.

## 1. Funkcja liniowa

### Przykład 1a: funkcja dodatnia

$$
f(x)=x+2 \quad \text{na } [0,2]
$$


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return x + 2

a, b = 0, 2
x = np.linspace(-1, 3, 400)
y = f(x)

integral_value, _ = quad(f, a, b)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='f(x)=x+2')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'Całka oznaczona = {integral_value:.4f}')
plt.legend()
plt.show()

print("Całka oznaczona:", integral_value)
````

### Przykład 1b: funkcja zmieniająca znak

$$
f(x)=x \quad \text{na } [-2,2]
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return x

a, b = -2, 2
x = np.linspace(-3, 3, 400)
y = f(x)

integral_value, _ = quad(f, a, b)
area_value, _ = quad(lambda t: abs(f(t)), a, b)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='f(x)=x')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'Całka = {integral_value:.4f}, pole figur = {area_value:.4f}')
plt.legend()
plt.show()

print("Całka oznaczona:", integral_value)
print("Pole figur:", area_value)
````

## 2. Funkcja kwadratowa

$$
f(x)=x^2 \quad \text{na } [0,2]
$$


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return x**2

a, b = 0, 2
x = np.linspace(-0.5, 2.5, 400)
y = f(x)

integral_value, _ = quad(f, a, b)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='f(x)=x^2')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'Całka oznaczona = {integral_value:.4f}')
plt.legend()
plt.show()

print("Całka oznaczona:", integral_value)
````

## 3. Funkcja Gaussa

### Standardowy rozkład normalny

$$
f(x)=\frac{1}{\sqrt{2\pi}}e^{-x^2/2}
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return 1 / np.sqrt(2 * np.pi) * np.exp(-x**2 / 2)

a, b = -1, 1
x = np.linspace(-4, 4, 600)
y = f(x)

integral_value, _ = quad(f, a, b)
total_area, _ = quad(f, -10, 10)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Gauss standardowy')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od -1 do 1:", integral_value)
print("Przybliżone całe pole:", total_area)
````


## 4. Rozkład jednostajny

Na $[0,4]$:

$$
f(x)=\frac14
$$


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return 0.25 if 0 <= x <= 4 else 0.0

a, b = 1, 3
x = np.linspace(-1, 5, 600)
y = np.array([f(t) for t in x])

integral_value, _ = quad(f, a, b)
total_area, _ = quad(f, -10, 10)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Rozkład jednostajny na [0,4]')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od 1 do 3:", integral_value)
print("Całe pole:", total_area)
````


## 5. Rozkład trójkątny

Na $[0,2]$ z maksimum w 1:

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    if 0 <= x <= 1:
        return x
    elif 1 < x <= 2:
        return 2 - x
    else:
        return 0.0

a, b = 0.5, 1.5
x = np.linspace(-0.5, 2.5, 600)
y = np.array([f(t) for t in x])

integral_value, _ = quad(f, a, b)
total_area, _ = quad(f, -10, 10)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Rozkład trójkątny')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od 0.5 do 1.5:", integral_value)
print("Całe pole:", total_area)
````

## 6. Rozkład wykładniczy

Dla $\lambda=1$:

$$
f(x)=e^{-x}, \quad x \ge 0
$$


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return np.exp(-x) if x >= 0 else 0.0

a, b = 1, 3
x = np.linspace(-1, 6, 600)
y = np.array([f(t) for t in x])

integral_value, _ = quad(f, a, b)
total_area, _ = quad(f, 0, 20)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Rozkład wykładniczy')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od 1 do 3:", integral_value)
print("Przybliżone całe pole:", total_area)
````

## 7. Rozkład Laplace’a

$$
f(x)=\frac12 e^{-|x|}
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return 0.5 * np.exp(-abs(x))

a, b = -1, 1
x = np.linspace(-6, 6, 600)
y = np.array([f(t) for t in x])

integral_value, _ = quad(f, a, b)
total_area, _ = quad(f, -20, 20)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Rozkład Laplace’a')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od -1 do 1:", integral_value)
print("Przybliżone całe pole:", total_area)
````

## 8. Rozkład Cauchy’ego

$$
f(x)=\frac{1}{\pi(1+x^2)}
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def f(x):
    return 1 / (np.pi * (1 + x**2))

a, b = -1, 1
x = np.linspace(-10, 10, 1000)
y = np.array([f(t) for t in x])

integral_value, _ = quad(f, a, b)
approx_area, _ = quad(f, -50, 50)

plt.figure(figsize=(7, 4))
plt.plot(x, y, label='Rozkład Cauchy’ego')
plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
plt.axhline(0)
plt.axvline(a, linestyle='--')
plt.axvline(b, linestyle='--')
plt.title(f'P({a} <= X <= {b}) = {integral_value:.4f}')
plt.legend()
plt.show()

print("Pole od -1 do 1:", integral_value)
print("Przybliżone całe pole:", approx_area)
````

## 9. Bardzo prosty szablon wielokrotnego użycia



````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def pokaż_funkcję_i_całkę(f, x_min, x_max, a, b, title):
    x = np.linspace(x_min, x_max, 800)
    y = np.array([f(t) for t in x])

    integral_value, _ = quad(f, a, b)

    plt.figure(figsize=(7, 4))
    plt.plot(x, y, label=title)
    plt.fill_between(x, y, where=(x >= a) & (x <= b), alpha=0.3)
    plt.axhline(0)
    plt.axvline(a, linestyle='--')
    plt.axvline(b, linestyle='--')
    plt.title(f'{title}, całka = {integral_value:.4f}')
    plt.legend()
    plt.show()

    print("Całka oznaczona:", integral_value)

def gauss(x):
    return 1 / np.sqrt(2 * np.pi) * np.exp(-x**2 / 2)

pokaż_funkcję_i_całkę(gauss, -4, 4, -1, 1, 'Gauss standardowy')
````
