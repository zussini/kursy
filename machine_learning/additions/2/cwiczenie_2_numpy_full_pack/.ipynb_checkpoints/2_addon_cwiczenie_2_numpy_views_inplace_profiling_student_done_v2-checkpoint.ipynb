## 0. Setup

W tej sekcji tworzymy dane testowe i helpery. Jeśli Twój komputer ma mało RAM, zmniejsz `N_BIG` (np. 50_000).

```python

import math
import numpy as np
from typing import Optional

RANDOM_SEED = 0
rng = np.random.default_rng(RANDOM_SEED)

# Rozmiary danych do ćwiczeń:
N_BIG = 200_000   # możesz zmniejszyć
D_BIG = 16
K_BIG = 16

def array_info(name, a):
    """Wypisz podstawowe informacje o tablicy."""
    print(f"{name}: shape={a.shape}, dtype={a.dtype}, nbytes={a.nbytes/1e6:.1f}MB, C_CONTIGUOUS={a.flags['C_CONTIGUOUS']}")

# Dane "duże" (float32 -> mniej RAM, typowy w ML)
X_big = rng.normal(size=(N_BIG, D_BIG), loc=0.0, scale=3.0).astype(np.float32)
C_big = rng.normal(size=(K_BIG, D_BIG), loc=0.0, scale=3.0).astype(np.float32)

array_info("X_big", X_big)
array_info("C_big", C_big)
```

## 1. View vs copy (nawyk: sprawdzaj, czy robisz kopię)

W NumPy **slice** zwykle zwraca *view* (bez kopii), a **fancy indexing** zwykle tworzy *copy*.

W praktyce na dużych danych to robi ogromną różnicę.

```python

def select_every_kth_col_view(X: np.ndarray, k: int) -> np.ndarray:
    """Zwróć view: co k‑tą kolumnę tablicy X (bez kopiowania)."""
    return X[:, ::k]


# tests ✅
X = np.arange(6*10, dtype=np.float32).reshape(6, 10)
Y = select_every_kth_col_view(X, 2)
assert Y.shape == (6, 5)
assert np.shares_memory(Y, X), "To ma być VIEW (bez kopii)."
Y[0, 0] = -123.0
assert X[0, 0] == -123.0
print("Ćwiczenie 1.1 tests passed ✅")
```

```python

def flatten_view_if_possible(X: np.ndarray) -> np.ndarray:
    """Zwróć 1D view danych, jeśli to możliwe."""
    return np.ravel(X)


# tests ✅
X = np.arange(12, dtype=np.float32).reshape(3, 4)  # C-contiguous
y = flatten_view_if_possible(X)
assert y.shape == (12,)
assert np.shares_memory(y, X), "Dla C-contiguous powinno dać się zrobić view."
y[0] = 999.0
assert X[0, 0] == 999.0
print("Ćwiczenie 1.2 tests passed ✅")
```

## 2. In‑place operacje (bez tworzenia kopii)

Tu ćwiczymy **`out=`** i operacje typu `X -= ...`, `X /= ...`.

**Cel nawykowy:** gdy pracujesz na dużych danych — najpierw myśl: *czy mogę to zrobić in‑place?*

```python

def clip_inplace(X: np.ndarray, lo: float, hi: float) -> np.ndarray:
    """Przytnij wartości X do [lo, hi] in-place."""
    np.clip(X, lo, hi, out=X)
    return X


# tests ✅
X = np.array([-10.0, -1.0, 0.0, 2.0, 100.0], dtype=np.float32)
X_id = id(X)
out = clip_inplace(X, -2.0, 3.0)
assert id(out) == X_id and out is X
assert float(out.min()) >= -2.0 - 1e-6
assert float(out.max()) <= 3.0 + 1e-6
print("Ćwiczenie 2.1 tests passed ✅")
```

```python

def center_columns_inplace(X: np.ndarray) -> np.ndarray:
    """Wycentruj kolumny X (odejmij mean) in-place."""
    mean = X.mean(axis=0, keepdims=True)
    X -= mean
    return X


# tests ✅
X = np.array([[1.0, 2.0], [3.0, 10.0], [5.0, 20.0]], dtype=np.float32)
X_id = id(X)
out = center_columns_inplace(X)
assert out is X and id(out) == X_id
assert np.allclose(out.mean(axis=0), 0.0, atol=1e-6)
print("Ćwiczenie 2.2 tests passed ✅")
```

```python

def standardize_columns_inplace(X: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    """Standaryzuj kolumny X in-place."""
    mean = X.mean(axis=0, keepdims=True)
    std = X.std(axis=0, keepdims=True)
    std = np.maximum(std, eps)
    X -= mean
    X /= std
    return X


# tests ✅
X = rng.normal(size=(1_000, 5), loc=10.0, scale=5.0).astype(np.float32)
X_id = id(X)
out = standardize_columns_inplace(X, eps=1e-6)
assert out is X and id(out) == X_id
assert np.all(np.isfinite(out))
assert np.allclose(out.mean(axis=0), 0.0, atol=2e-2)
assert np.allclose(out.std(axis=0), 1.0, atol=2e-2)
print("Ćwiczenie 2.3 tests passed ✅")
```

## 3. Sprawdzanie poprawności funkcji (kontrakty)

W ML łatwo zrobić błąd, który *nie wywala wyjątku*, ale psuje wyniki.

Ćwiczymy nawyk: **pisz małe testy/kontrakty** (asserty), zanim pójdziesz dalej.

```python

def check_standardized(X: np.ndarray, atol_mean: float = 1e-2, atol_std: float = 1e-2) -> bool:
    assert isinstance(X, np.ndarray)
    assert np.all(np.isfinite(X)), "X contains NaN/inf"
    m = X.mean(axis=0)
    s = X.std(axis=0)
    assert np.allclose(m, 0.0, atol=atol_mean), f"mean not ~0: {m}"
    assert np.allclose(s, 1.0, atol=atol_std), f"std not ~1: {s}"
    return True


# tests ✅
X = rng.normal(size=(10_000, 3)).astype(np.float32)
X_bad = X.copy()
X_good = X.copy()
standardize_columns_inplace(X_good, eps=1e-6)

assert check_standardized(X_good, atol_mean=2e-2, atol_std=2e-2) is True

failed = False
try:
    check_standardized(X_bad, atol_mean=2e-2, atol_std=2e-2)
except AssertionError:
    failed = True
assert failed
print("Ćwiczenie 3.1 tests passed ✅")
```

## 4. Case study: szybkie liczenie odległości (k‑means / k‑NN)

Mamy punkty `X` (N×D), centroidy `C` (K×D) i chcemy policzyć macierz:
`dist2[i, k] = ||X[i] - C[k]||^2` o kształcie (N×K).

### Trik (ważny w praktyce!)
\[
\|x-c\|^2 = \|x\|^2 + \|c\|^2 - 2 x\cdot c
\]

Dzięki temu:
- unikamy pętli w Pythonie,
- unikamy wielkiej tablicy 3D (N×K×D),
- przenosimy pracę do szybkich rutyn BLAS (mnożenie macierzy).

```python

# Funkcja bazowa (wolna) — gotowa. Służy do sprawdzenia poprawności na małej próbce.

def squared_euclidean_distances_slow(X: np.ndarray, C: np.ndarray) -> np.ndarray:
    """Wolna referencja: dist2[i, k] = ||X[i]-C[k]||^2."""
    N, D = X.shape
    K, D2 = C.shape
    assert D == D2
    out = np.empty((N, K), dtype=np.float64)
    for i in range(N):
        for k in range(K):
            diff = X[i] - C[k]
            out[i, k] = float(np.dot(diff, diff))
    return out
```

```python

def squared_euclidean_distances_fast(X: np.ndarray, C: np.ndarray, out: Optional[np.ndarray] = None) -> np.ndarray:
    """Szybka wersja: ||x-c||^2 = ||x||^2 + ||c||^2 - 2 x·c."""
    X = np.asarray(X)
    C = np.asarray(C)
    N, D = X.shape
    K, D2 = C.shape
    assert D == D2

    if out is None:
        out = X @ C.T
    else:
        np.matmul(X, C.T, out=out)

    out *= -2.0
    x_norm2 = np.sum(X * X, axis=1, keepdims=True)                 # (N,1)
    c_norm2 = np.sum(C * C, axis=1, keepdims=True).T               # (1,K)
    out += x_norm2
    out += c_norm2
    np.maximum(out, 0.0, out=out)
    return out


# tests ✅ (małe dane)
X = rng.normal(size=(200, 4)).astype(np.float64)
C = rng.normal(size=(5, 4)).astype(np.float64)

d_slow = squared_euclidean_distances_slow(X, C)
d_fast = squared_euclidean_distances_fast(X, C)

assert d_fast.shape == d_slow.shape
assert np.allclose(d_fast, d_slow, atol=1e-10)
assert np.all(d_fast >= -1e-9)
print("Ćwiczenie 4.1 tests passed ✅")
```

```python

def assign_nearest_centroid(dist2: np.ndarray) -> np.ndarray:
    return np.argmin(dist2, axis=1)


# tests ✅
dist2 = np.array([[1.0, 2.0, 0.5],
                  [9.0, 3.0, 4.0]], dtype=np.float64)
idx = assign_nearest_centroid(dist2)
assert idx.shape == (2,)
assert np.array_equal(idx, np.array([2, 1]))
print("Ćwiczenie 4.2 tests passed ✅")
```

### 4.3 (bonus) — minimalizuj kopie: pracuj na *view* wybranych cech

Czasem chcesz liczyć odległości tylko na części cech (np. pierwsze D' kolumn).
Zróbmy to tak, żeby **nie kopiować** danych: użyj slice (view).

```python

def first_d_features_view(X: np.ndarray, d: int) -> np.ndarray:
    return X[:, :d]


# tests ✅
X = rng.normal(size=(1000, 10)).astype(np.float32)
C = rng.normal(size=(7, 10)).astype(np.float32)

X_sub = first_d_features_view(X, 4)
C_sub = first_d_features_view(C, 4)

assert X_sub.shape == (1000, 4)
assert C_sub.shape == (7, 4)
assert np.shares_memory(X_sub, X)
assert np.shares_memory(C_sub, C)

d1 = squared_euclidean_distances_fast(X_sub, C_sub)
d2 = squared_euclidean_distances_fast(X_sub.copy(), C_sub.copy())
assert np.allclose(d1, d2, atol=1e-6)
print("Ćwiczenie 4.3 tests passed ✅")
```

## 5. Profilowanie: `%prun` i `%lprun`

W tej sekcji *używasz narzędzi* z IPython, żeby zobaczyć gdzie jest wąskie gardło.

Uwaga: wyniki zależą od sprzętu — nie oceniamy liczb, tylko **to, czy potrafisz wyciągnąć wniosek**.

```python

N_PRUN = 2000
D_PRUN = 16
K_PRUN = 8

X_prun = rng.normal(size=(N_PRUN, D_PRUN)).astype(np.float64)
C_prun = rng.normal(size=(K_PRUN, D_PRUN)).astype(np.float64)

ANSWER_PRUN = (
    "W %prun widać, że większość czasu spędzamy w podwójnej pętli oraz w operacjach wewnątrz pętli "
    "(diff = X[i]-C[k] oraz dot(diff,diff)). To sugeruje, że wąskim gardłem jest Python loop, a nie sam NumPy. "
    "Rozwiązaniem jest wektoryzacja: policzyć macierz iloczynów X@C.T oraz dodać normy (trik algebraiczny), "
    "co eliminuje pętle w Pythonie i przenosi pracę do zoptymalizowanych rutyn BLAS."
)

assert isinstance(ANSWER_PRUN, str)
assert len(ANSWER_PRUN.strip()) >= 200
print("Profilowanie (%prun) — odpowiedź uzupełniona ✅")
```

```python

ANSWER_LPRUN = (
    "Najwięcej czasu idzie w linie wewnątrz pętli: diff = X[i]-C[k] oraz dot(diff,diff). "
    "To potwierdza, że problemem jest pętla w Pythonie i powtarzane operacje dla każdego (i,k). "
    "Rozwiązaniem jest wektoryzacja (X@C.T i normy) albo batchowanie."
)

assert isinstance(ANSWER_LPRUN, str)
assert len(ANSWER_LPRUN.strip()) >= 10
print("Profilowanie (%lprun) — OK ✅")
```

## 5.2 Benchmark: `%timeit -o` i *speedup* (nawyk po profilowaniu)

Skoro już wiesz (z `%prun` / `%lprun`) **gdzie** jest wąskie gardło, zróbmy krok dalej:
zmierzmy **czas całej funkcji** i policzmy *speedup* po optymalizacji.

**Cel ćwiczenia:** wyrobić nawyk:
1) najpierw *poprawność* (testy),  
2) potem *profilowanie* (gdzie idzie czas),  
3) na końcu *benchmark* (czy poprawa jest realna).

**Wskazówki:**
- Użyj `%timeit -o`, żeby zapisać wynik do zmiennej (`t_slow`, `t_fast`).
- Użyj **stałych** `-n` i `-r` (wtedy czas będzie przewidywalny na zajęciach).
- Jeśli wersja wolna trwa za długo, zmniejsz `N_TIMEIT` (np. 200).


```python
# Przygotuj dane do benchmarku
N_TIMEIT = 300
D_TIMEIT = 16
K_TIMEIT = 8

X_timeit = rng.normal(size=(N_TIMEIT, D_TIMEIT)).astype(np.float64)
C_timeit = rng.normal(size=(K_TIMEIT, D_TIMEIT)).astype(np.float64)

# Benchmark: wolno vs szybko (stałe -n i -r, żeby było przewidywalne na zajęciach)
t_slow = %timeit -n 3 -r 3 -o squared_euclidean_distances_slow(X_timeit, C_timeit)
t_fast = %timeit -n 3 -r 3 -o squared_euclidean_distances_fast(X_timeit, C_timeit)

speedup = float(t_slow.average / t_fast.average)

ANSWER_TIMEIT = (
    "Wersja szybka jest znacząco szybsza, bo eliminuje pętle w Pythonie i przenosi obliczenia do "
    "wektorowych operacji NumPy/BLAS (X @ C.T + normy), a także unika tworzenia wielu małych tablic "
    "w środku pętli. %timeit pokazuje realny zysk end-to-end (speedup)."
)
```

```python
# tests ✅ (czasów nie sprawdzamy na sztywno — tylko typy i podstawowe relacje)
assert t_slow is not None and hasattr(t_slow, "average"), "Uruchom %timeit -o dla wersji wolnej (t_slow)"
assert t_fast is not None and hasattr(t_fast, "average"), "Uruchom %timeit -o dla wersji szybkiej (t_fast)"
assert isinstance(speedup, (float, int)) and speedup > 1.0, "speedup powinien być > 1.0 (wolna wersja ma być wolniejsza)"
assert len(ANSWER_TIMEIT.strip()) >= 120, "Uzupełnij ANSWER_TIMEIT (min. 120 znaków)"
print(f"Benchmark ok ✅  speedup ≈ {speedup:.2f}x")
```

## 6. Bonus: policz, ile pamięci zjadłaby wersja 3D (broadcasting)

Wersja:
`(X[:,None,:] - C[None,:,:])**2` tworzy tablicę (N×K×D).

Policzmy, ile to by było w MB/GB dla danych podobnych do `X_big`.

```python

def bytes_for_shape(shape, dtype=np.float32):
    return int(np.prod(shape)) * np.dtype(dtype).itemsize

bytes_3d = bytes_for_shape((N_BIG, K_BIG, D_BIG), dtype=np.float32)
mb_3d = bytes_3d / 1e6
gb_3d = bytes_3d / 1e9

ANSWER_MEMORY = (
    f"Dla N={N_BIG}, K={K_BIG}, D={D_BIG} 3D tablica miałaby ok. {gb_3d:.2f} GB. "
    "To często jest nie do przełknięcia na laptopie, a dodatkowo tworzenie takiej tablicy jest kosztowne czasowo. "
    "Dlatego lepiej użyć triku z normami i iloczynem macierzy, który unika alokacji (N,K,D) i operuje na (N,K) "
    "+ małe wektory norm."
)

assert gb_3d > 0
assert len(ANSWER_MEMORY.strip()) >= 120
print(f"3D intermediate: ~{mb_3d:.1f} MB (~{gb_3d:.2f} GB) ✅")
```
