### Ćwiczenie 1.3 — Sygnatura funkcji i pierwsza linia docstringa

### Jak rozwiązać to zadanie (krok po kroku)

Twoim celem jest szybkie podejrzenie:
- **jak wywołać funkcję** (argumenty),
- **co ona robi** (krótki opis),

bez czytania całej dokumentacji.

Zrób to tak:

1. **Weź funkcję**, np. `train_test_split`.

2. **Sprawdź jej sygnaturę**:
   ```python
   import inspect
   inspect.signature(train_test_split)

````python

def get_signature_str(obj) -> str:
    return str(inspect.signature(obj))

def get_first_doc_line(obj) -> str:
    doc = inspect.getdoc(obj) or ""
    for line in doc.splitlines():
        if line.strip():
            return line.strip()
    return ""
````

````python

from sklearn.model_selection import train_test_split

sig = get_signature_str(train_test_split)
assert isinstance(sig, str) and "test_size" in sig and "random_state" in sig

first = get_first_doc_line(train_test_split)
assert_nonempty_str(first, min_len=10)
ok("Ćwiczenie 1.3")
````

````python
print(str(inspect.signature(train_test_split)))
#print(inspect.getdoc(train_test_split))
doc = inspect.getdoc(train_test_split)
first_line = doc.splitlines()[0]
print(first_line)
````

### Ćwiczenie 1.5 — Eksploracja modułu (wildcard na `dir()`)

````python

def find_public_names(module, pattern: str) -> list[str]:
    names = [n for n in dir(module) if not n.startswith("_")]
    return sorted([n for n in names if fnmatch.fnmatch(n, pattern)])

RANDOM_INT_FUNCTION_NAME = "randint"
````

````python

names = find_public_names(np.random, "*rand*")
assert isinstance(names, list) and len(names) > 0

assert isinstance(RANDOM_INT_FUNCTION_NAME, str) and RANDOM_INT_FUNCTION_NAME
assert hasattr(np.random, RANDOM_INT_FUNCTION_NAME)
fn = getattr(np.random, RANDOM_INT_FUNCTION_NAME)
sample = fn(0, 10, size=5)
assert np.asarray(sample).dtype.kind in {"i","u"}
ok("Ćwiczenie 1.5")
````

````python
import fnmatch

def find_public_names(module, pattern: str) -> list[str]:
    return [
        name for name in dir(module)
        if not name.startswith("_") and fnmatch.fnmatch(name, pattern)
    ]
````

### Ćwiczenie 3.1 — Policz magics (line vs cell)

````python

def count_magics() -> tuple[int,int]:
    ip = get_ipython()
    magics = ip.magics_manager.lsmagic()
    return len(magics.get("line", {})), len(magics.get("cell", {}))
````

### Ćwiczenie 3.3 — `writefile` + `%run` (skrypt)

````python

def write_myscript(path: Path) -> None:
    lines = [
        "def square(x):",
        "    '''square a number'''",
        "    return x ** 2",
        "",
        "if __name__ == '__main__':",
        "    for N in range(1, 4):",
        "        print(N, 'squared is', square(N))",
        "",
    ]
    path.write_text("\n".join(lines), encoding="utf-8")

SCRIPT_PATH = SANDBOX_DIR / "myscript_student.py"
````

````python

write_myscript(SCRIPT_PATH)
assert SCRIPT_PATH.exists()

ip = get_ipython()
ip.run_line_magic("run", str(SCRIPT_PATH))
assert "square" in globals()
assert square(6) == 36
ok("Ćwiczenie 3.3")
````

### Ćwiczenie 3.4 — `%timeit`: pętla vs NumPy

````python

def sum_squares_loop(n: int) -> int:
    s = 0
    for i in range(n):
        s += i*i
    return s

def sum_squares_vec(n: int) -> int:
    x = np.arange(n, dtype=np.int64)
    return int(np.sum(x*x))
````

````python

assert sum_squares_loop(10_000) == sum_squares_vec(10_000)
ok("Ćwiczenie 3.4")
````

## 01.04 — Input / Output History

Cel: `_`, `__`, `___` oraz podstawowe operacje na historii wejść.

### Ćwiczenie 4.1 — `_`, `__`, `___` (w tej samej komórce!)

````python
def capture_underscores():
    ip = get_ipython()
    ip.run_cell("2 + 2", store_history=True)
    ip.run_cell("'abra' * 2", store_history=True)
    ip.run_cell("sum([1, 2, 3])", store_history=True)
    return _, __, ___


LAST, PREV, PREV2 = capture_underscores()
````

````python

assert LAST == 6 and PREV == "abraabra" and PREV2 == 4
ok("Ćwiczenie 4.1")
````

### Ćwiczenie 4.3 — `history_manager.get_tail`

````python

def tail_inputs(n: int) -> list[str]:
    ip = get_ipython()
    rows = ip.history_manager.get_tail(n, raw=True, output=False)
    return [inp for (_sess,_line,inp) in rows]
````

````python

ip = get_ipython()
ip.run_cell("HISTMARK_VAR = 123  # HISTMARK_ABC", store_history=True)
tail = tail_inputs(5)
assert any("HISTMARK_ABC" in x for x in tail)
ok("Ćwiczenie 4.3")
````

### Ćwiczenie 4.4 — `history_manager.search`

````python

def search_history(pattern: str) -> list[str]:
    ip = get_ipython()
    rows = ip.history_manager.search(pattern, raw=True, output=False)
    return [inp for (_sess,_line,inp) in rows]
````

````python

found = search_history("*HISTMARK_ABC*")
assert isinstance(found, list) and any("HISTMARK_ABC" in x for x in found)
ok("Ćwiczenie 4.4")
````

### Ćwiczenie 4.5 — Zapisz ostatnie wejścia do pliku (mini-`%save`)

````python

def save_last_inputs(path: Path, n: int) -> None:
    lines = tail_inputs(n)
    path.write_text("\n".join(lines) + "\n", encoding="utf-8")

HISTORY_SAVE_PATH = SANDBOX_DIR / "saved_history_snippet.py"
````

````python

save_last_inputs(HISTORY_SAVE_PATH, 10)
assert HISTORY_SAVE_PATH.exists()
assert "HISTMARK_ABC" in HISTORY_SAVE_PATH.read_text(encoding="utf-8")
ok("Ćwiczenie 4.5")
````

## 01.05 — IPython i komendy shellowe

### Ćwiczenie 5.1 — Listing katalogu (IPython + fallback)

````python

def list_dir(path: Path) -> list[str]:
    ip = get_ipython()
    if os.name == "nt":
        out = ip.getoutput(f'dir /b "{path}"')
    else:
        out = ip.getoutput(f'ls "{path}"')
    out = [x.strip() for x in out if x.strip()]
    if out:
        return out
    return sorted([p.name for p in path.iterdir()])
````

````python

listing = list_dir(SANDBOX_DIR)
assert "hello.txt" in listing and "mini.csv" in listing
ok("Ćwiczenie 5.1")
````

### Ćwiczenie 5.4 — Zmiana katalogu (odpowiednik `%cd`)

````python

def cd_and_back(target: Path) -> tuple[Path,Path,Path]:
    ip = get_ipython()
    orig = Path.cwd()
    ip.run_line_magic("cd", str(target))
    inside = Path.cwd()
    ip.run_line_magic("cd", str(orig))
    back = Path.cwd()
    return orig, inside, back
````

````python

orig, inside, back = cd_and_back(SANDBOX_DIR)
assert inside.resolve() == SANDBOX_DIR.resolve()
assert back.resolve() == orig.resolve()
ok("Ćwiczenie 5.4")
````
