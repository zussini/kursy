# `reshape`, `ravel` i `meshgrid`

Samodzielny zestaw wprawek do mostu NumPy -> wizualizacja. Celem jest zrozumienie ksztaltow tablic przed rysowaniem powierzchni, konturow i punktow siatki.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Ksztalt tablicy

```python
values = np.arange(12)
matrix = values.reshape(3, 4)

print(values)
print(matrix)
print(matrix.shape)
```

`ravel()` splaszcza tablice. Przydaje sie, gdy funkcja rysujaca oczekuje listy punktow zamiast macierzy wspolrzednych.

```python
flat = matrix.ravel()
print(flat)
print(flat.shape)
```

## `meshgrid` jako generator punktow

```python
x = np.linspace(-2, 2, 5)
y = np.linspace(-1, 1, 3)

X, Y = np.meshgrid(x, y)
print("X:\n", X)
print("Y:\n", Y)
print(X.shape, Y.shape)
```

```python
points = np.column_stack([X.ravel(), Y.ravel()])
points[:10]
```

Ten sam zestaw punktow mozna narysowac jako zwykly scatter plot.

```python
plt.figure(figsize=(5, 3))
plt.scatter(points[:, 0], points[:, 1], s=80)
plt.axhline(0, color="0.8", linewidth=1)
plt.axvline(0, color="0.8", linewidth=1)
plt.gca().set_aspect("equal", adjustable="box")
plt.title("Punkty utworzone przez meshgrid")
plt.xlabel("x")
plt.ylabel("y");
```

## Wartosc funkcji na siatce

```python
Z = np.sin(X) * np.cos(Y)
print(Z.shape)
Z
```

```python
plt.figure(figsize=(5, 3))
plt.contourf(X, Y, Z, levels=20, cmap="viridis")
plt.colorbar(label="sin(x) cos(y)")
plt.scatter(points[:, 0], points[:, 1], color="white", edgecolor="black")
plt.title("Funkcja policzona na siatce")
plt.xlabel("x")
plt.ylabel("y");
```
