## Cel
To jest główny blok techniczny.  
Ćwiczymy dokładnie to, co wyszło najsłabiej:
- filtrowanie,
- braki danych,
- `groupby`,
- `merge`,
- `pivot_table`.


```python
import pandas as pd
import numpy as np

results = pd.DataFrame({
    "runner_id": [1, 2, 3, 4, 5, 6],
    "gender": ["K", "M", "K", "M", "K", "M"],
    "age_group": ["25-29", "25-29", "30-34", "30-34", "35-39", "35-39"],
    "final_sec": [7600, 7300, 7450, 7200, np.nan, 7500],
})

clubs = pd.DataFrame({
    "runner_id": [1, 2, 3, 4, 5, 6],
    "club": ["A", "A", "B", "B", "A", "C"]
})

results
```

## Zadanie 1. Filtrowanie
Wybierz:
- tylko kobiety,
- tylko grupę `30-34`,
- tylko osoby z czasem poniżej 7500 sekund.


```python
women = results[results["gender"] == "K"]
age_30_34 = results[results["age_group"] == "30-34"]
fast = results[results["final_sec"] < 7500]

women, age_30_34, fast
```

## Zadanie 2. Braki danych
Sprawdź:
- gdzie są braki danych,
- ile ich jest,
- usuń lub uzupełnij brak czasu końcowego średnią z kolumny.


```python
results.isna()
```

```python
results["final_sec"].isna().sum(), results["final_sec"].mean()
```

```python
results_filled = results.copy()
results_filled["final_sec"] = results_filled["final_sec"].fillna(results_filled["final_sec"].mean())
results_filled
```

## Zadanie 3. `groupby`
Policz średni czas końcowy:
- po `age_group`,
- po `gender`,
- po `gender` i `age_group`.


```python
results_filled.groupby("age_group")["final_sec"].mean()
```

```python
results_filled.groupby("gender")["final_sec"].mean()
```

```python
results_filled.groupby(["gender", "age_group"])["final_sec"].mean()
```

## Zadanie 4. `merge`
Połącz tabelę wyników z tabelą klubów.


```python
merged = results_filled.merge(clubs, on="runner_id", how="left")
merged
```

## Zadanie 5. `pivot_table`
Zrób tabelę przestawną:
- wiersze = `age_group`,
- kolumny = `gender`,
- wartości = średni `final_sec`.


```python
pivot = pd.pivot_table(
    merged,
    index="age_group",
    columns="gender",
    values="final_sec",
    aggfunc="mean"
)
pivot
```

## Zadanie 6. Stringi
Dodaj kolumnę `age_min`, która wyciąga dolną granicę wieku z `age_group`.


```python
merged["age_min"] = merged["age_group"].str.split("-").str[0].astype(int)
merged
```
