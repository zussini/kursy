## Zadanie 5. Krótka wprawka OOP
Utwórz klasę `StudentProfile` z polami:
- `name`,
- `python_score`,
- `data_score`.

Dodaj metodę `needs_support()`, która zwraca `True`, jeśli średnia z dwóch wyników < 3.


```python
class StudentProfile:
    def __init__(self, name, python_score, data_score):
        self.name = name
        self.python_score = python_score
        self.data_score = data_score

    def needs_support(self):
        return (self.python_score + self.data_score) / 2 < 3

s = StudentProfile("Ala", 4, 2)
s.needs_support()
```
