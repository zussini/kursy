## Cel
Skupiamy się tylko na tym, co w ankiecie wyszło słabiej:
- `shape`,

- różnica między pętlą a podejściem wektorowym.


```python
import numpy as np
```

## Zadanie 1. Kształt danych
Utwórz tablicę 2D z wynikami:
- wiersze = zawodnicy,
- kolumny = `split_5k`, `split_10k`, `final`.

Sprawdź `shape`, `ndim`, `size`.


```python
times = np.array([
    [1500, 3200, 7600],
    [1450, 3050, 7300],
    [1600, 3400, 8100],
    [1550, 3300, 7900],
])
times.shape, times.ndim, times.size
```

## Zadanie 2. Indeksowanie i slicing
Wyciągnij:
- pierwszy wiersz,
- ostatnią kolumnę,
- pierwsze dwa wiersze i dwie pierwsze kolumny.


```python
first_runner = times[0]
final_times = times[:, -1]
sub = times[:2, :2]

first_runner, final_times, sub
```

## Zadanie 3. Maski
Zaznacz zawodników z czasem końcowym poniżej 7800 sekund.


```python
mask_fast = times[:, -1] < 7800
mask_fast, times[mask_fast]
```

## Zadanie 4. Agregacje
Policz:
- średni czas końcowy,
- minimalny czas końcowy,
- średni czas końcowy tylko dla zawodników z maski `mask_fast`.


```python
final_times = times[:, -1]
final_times.mean(), final_times.min(), final_times[mask_fast].mean()
```

## Zadanie 5. Pętla vs wektorowość
Policzmy poprawiony wynik `time * 0.98`.

1. zrób to pętlą po liście,
2. zrób to wektorowo na NumPy,
3. porównaj zapis.


```python
times_list = [7600, 7300, 8100, 7900]
improved_loop = []
for t in times_list:
    improved_loop.append(t * 0.98)

improved_np = final_times * 0.98

improved_loop, improved_np
```
