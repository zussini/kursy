### groupby() w Pandas

Metoda `groupby()` służy do dzielenia danych na grupy według wartości w wybranej kolumnie, a następnie do wykonywania obliczeń osobno dla każdej grupy.

Dzięki temu można szybko porównywać różne kategorie danych, np. policzyć liczbę elementów, średnią, medianę lub sumę w każdej grupie.

Przykład:
- grupowanie uczniów według tego, czy zdali,
- grupowanie produktów według kategorii,
- grupowanie rekordów według miasta lub płci.

### Po co używamy groupby()?

`groupby()` dzieli dane na grupy według wartości wybranej kolumny lub kilku kolumn. Następnie możemy dla każdej grupy osobno policzyć liczbę obserwacji, średnią, medianę, minimum, maksimum lub inne statystyki.

Metoda groupby() w bibliotece Pandas (Python) służy do grupowania danych w ramce danych (DataFrame) na podstawie wartości w jednej lub kilku kolumnach. Działa analogicznie do klauzuli GROUP BY w języku SQL. 

- .sum() – Sumowanie wartości w grupach.
- .mean() – Średnia arytmetyczna grup.
- .count() / .size() – Liczenie wierszy w każdej grupie.
- .median() – Mediana.
- .agg() – Zastosowanie różnych funkcji do różnych kolumn jednocześnie.

## boxploty


## Boxplot

Boxplot pokazuje medianę, kwartyle ([Q1,Q2,Q3]-50% danych,Q4) i ewentualne wartości odstające. Jest dobrym, zwartym podsumowaniem rozkładu danych.

Boxplot pokazuje medianę, pierwszy i trzeci kwartyl oraz ewentualne wartości odstające. Pudełko obejmuje środkowe 50% danych, a wąsy pokazują dalszy rozrzut wartości.

## Kwartyle – co oznaczają

5 kluczowych wartości (tzw. **five-number summary**):

1. **Minimum (min)**
   → najmniejsza wartość w danych

2. **Q1 (pierwszy kwartyl)**
   → 25% danych jest **mniejszych lub równych** tej wartości

3. **Q2 (mediana)**
   → 50% danych jest poniżej, 50% powyżej
   → to jest **środek rozkładu**

4. **Q3 (trzeci kwartyl)**
   → 75% danych jest **mniejszych lub równych** tej wartości

5. **Maximum (max)**
   → największa wartość w danych

---

## Intuicja

Wyobraź sobie dane posortowane rosnąco:

```
min ---- Q1 ---- median (Q2) ---- Q3 ---- max
```

To dzieli dane na 4 równe części:

* od **min do Q1** → 25% danych
* od **Q1 do mediany** → kolejne 25%
* od **mediany do Q3** → kolejne 25%
* od **Q3 do max** → ostatnie 25%

---

* **Q1 to NIE minimum**
* **Q3 to NIE maksimum**

Tylko:

* min → najmniejsza wartość
* Q1 → 25% punkt
* median (Q2) → 50% punkt
* Q3 → 75% punkt
* max → największa wartość

---

## Jak to wygląda w boxplacie

* dolna kreska → **min**
* dolna krawędź pudełka → **Q1**
* linia w środku → **mediana**
* górna krawędź pudełka → **Q3**
* górna kreska → **max**

(ew. bez outlierów, bo czasem whiskery są obcięte do 1.5 IQR)

---

## Mini przykład liczbowy

Dane:

```
[1, 2, 3, 4, 5, 6, 7, 8]
```

* min = 1
* Q1 ≈ 2.5
* mediana = 4.5
* Q3 ≈ 6.5
* max = 8

---

## Stworzenie funkcji, którą możemy potem wykorzystywać do pokazywania rozkładu z wykorzystaniem wcześniejszych metod

## Automatyzacja powtarzalnej analizy

Jeśli kilka razy wykonujemy podobne obliczenia i rysujemy podobne wykresy, warto zamknąć ten kod w funkcji. Dzięki temu analiza staje się krótsza i bardziej uporządkowana.

````python
# Create a function that we can re-use
def show_distribution(var_data):
    from matplotlib import pyplot as plt

    # Get statistics
    min_val = var_data.min()
    max_val = var_data.max()
    mean_val = var_data.mean()
    med_val = var_data.median()
    mod_val = var_data.mode()[0]

    print('Minimum:{:.2f}\nMean:{:.2f}\nMedian:{:.2f}\nMode:{:.2f}\nMaximum:{:.2f}\n'.format(min_val,
                                                                                            mean_val,
                                                                                            med_val,
                                                                                            mod_val,
                                                                                            max_val))

    # Create a figure for 2 subplots (2 rows, 1 column)
    fig, ax = plt.subplots(2, 1, figsize = (10,4))

    # Plot the histogram   
    ax[0].hist(var_data)
    ax[0].set_ylabel('Frequency')

    # Add lines for the mean, median, and mode
    ax[0].axvline(x=min_val, color = 'gray', linestyle='dashed', linewidth = 2)
    ax[0].axvline(x=mean_val, color = 'cyan', linestyle='dashed', linewidth = 2)
    ax[0].axvline(x=med_val, color = 'red', linestyle='dashed', linewidth = 2)
    ax[0].axvline(x=mod_val, color = 'yellow', linestyle='dashed', linewidth = 2)
    ax[0].axvline(x=max_val, color = 'gray', linestyle='dashed', linewidth = 2)

    # Plot the boxplot   
    ax[1].boxplot(var_data, vert=False)
    ax[1].set_xlabel('Value')

    # Add a title to the Figure
    fig.suptitle('Data Distribution')

    # Show the figure
    fig.show()

# Get the variable to examine
col = df_students['Grade']
# Call the function
show_distribution(col)
````

## PDF - probability density function - estymacja ogólnej populacji - całkowanie/uciąglenie rozkładu (bell_curve)

## Uwaga

Wykres gęstości jest wygładzonym przybliżeniem rozkładu danych. Nie pokazuje dokładnych częstości jak histogram, tylko estymację ciągłej gęstości.

````python
def show_density(data):
    fig = plt.figure(figsize=(10,4))
    data.plot.density()
    plt.title('Data Density')
    plt.show()
show_density(df_students.Grade)
````

````python
def show_density(var_data):
    from matplotlib import pyplot as plt

    fig = plt.figure(figsize=(10,4))

    # Plot density
    var_data.plot.density()

    # Add titles and labels
    plt.title('Data Density')

    # Show the mean, median, and mode
    
    plt.axvline(x=var_data.mean(), color = 'cyan', linestyle='dashed', linewidth = 2)
    plt.axvline(x=var_data.median(), color = 'red', linestyle='dashed', linewidth = 2)
    plt.axvline(x=var_data.mode()[0], color = 'yellow', linestyle='dashed', linewidth = 2)
    #plt.axvline(x=min_val, color = 'gray', linestyle='dashed', linewidth = 2)
    #plt.axvline(x=max_val, color = 'gray', linestyle='dashed', linewidth = 2)

    # Show the figure
    plt.show()

# Get the density of Grade
col = df_students['Grade']
show_density(col)
````

````python
import itertools
import numpy as np
import pandas as pd
import plotly.graph_objects as go

# --------------------------------------------------
# 1. Przykładowe dane
# --------------------------------------------------
np.random.seed(42)

genders = ['F', 'M']
cities = ['Krakow', 'Warsaw']
passed = ['No', 'Yes']

rows = []
for g, c, p in itertools.product(genders, cities, passed):
    n = np.random.randint(8, 16)
    for _ in range(n):
        rows.append({
            'Gender': g,
            'City': c,
            'Passed': p,
            'ExamScore': np.random.normal(
                loc=75 if p == 'Yes' else 45,
                scale=10
            )
        })

df = pd.DataFrame(rows)
df['ExamScore'] = df['ExamScore'].clip(0, 100)

# --------------------------------------------------
# 2. Mapowanie kategorii na osie sześcianu
# --------------------------------------------------
x_map = {'F': 0, 'M': 1}
y_map = {'Krakow': 0, 'Warsaw': 1}
z_map = {'No': 0, 'Yes': 1}

df['x0'] = df['Gender'].map(x_map)
df['y0'] = df['City'].map(y_map)
df['z0'] = df['Passed'].map(z_map)

# Punkty losowo rozrzucone wewnątrz każdej komórki 1x1x1
margin = 0.12
df['x'] = df['x0'] + np.random.uniform(margin, 1 - margin, len(df))
df['y'] = df['y0'] + np.random.uniform(margin, 1 - margin, len(df))
df['z'] = df['z0'] + np.random.uniform(margin, 1 - margin, len(df))

# --------------------------------------------------
# 3. Funkcja do rysowania krawędzi prostopadłościanu
# --------------------------------------------------
def cuboid_edges(x0, x1, y0, y1, z0, z1):
    corners = np.array([
        [x0, y0, z0],
        [x1, y0, z0],
        [x1, y1, z0],
        [x0, y1, z0],
        [x0, y0, z1],
        [x1, y0, z1],
        [x1, y1, z1],
        [x0, y1, z1],
    ])

    edges = [
        (0,1), (1,2), (2,3), (3,0),
        (4,5), (5,6), (6,7), (7,4),
        (0,4), (1,5), (2,6), (3,7)
    ]

    traces = []
    for i, j in edges:
        traces.append(go.Scatter3d(
            x=[corners[i,0], corners[j,0]],
            y=[corners[i,1], corners[j,1]],
            z=[corners[i,2], corners[j,2]],
            mode='lines',
            line=dict(width=5),
            showlegend=False,
            hoverinfo='skip'
        ))
    return traces

# --------------------------------------------------
# 4. Rysowanie
# --------------------------------------------------
fig = go.Figure()

# Punkty, kolorowane wynikiem egzaminu
fig.add_trace(go.Scatter3d(
    x=df['x'],
    y=df['y'],
    z=df['z'],
    mode='markers',
    marker=dict(
        size=5,
        color=df['ExamScore'],
        colorscale='Viridis',
        colorbar=dict(title='ExamScore'),
        opacity=0.85
    ),
    text=[
        f"Gender={g}<br>City={c}<br>Passed={p}<br>Score={s:.1f}"
        for g, c, p, s in zip(df['Gender'], df['City'], df['Passed'], df['ExamScore'])
    ],
    hovertemplate='%{text}<extra></extra>',
    name='Students'
))

# Krawędzie dużego sześcianu 2x2x2
for tr in cuboid_edges(0, 2, 0, 2, 0, 2):
    tr.line.color = 'black'
    fig.add_trace(tr)

# Wewnętrzne podziały grup
inner_lines = [
    ([1,1], [0,2], [0,0]), ([1,1], [0,2], [2,2]),
    ([1,1], [0,0], [0,2]), ([1,1], [2,2], [0,2]),
    ([0,2], [1,1], [0,0]), ([0,2], [1,1], [2,2]),
    ([0,0], [1,1], [0,2]), ([2,2], [1,1], [0,2]),
    ([0,0], [0,2], [1,1]), ([2,2], [0,2], [1,1]),
    ([0,2], [0,0], [1,1]), ([0,2], [2,2], [1,1]),
]
for xs, ys, zs in inner_lines:
    fig.add_trace(go.Scatter3d(
        x=xs, y=ys, z=zs,
        mode='lines',
        line=dict(color='gray', width=3, dash='dash'),
        showlegend=False,
        hoverinfo='skip'
    ))

fig.update_layout(
    title='Wielowymiarowe grupowanie danych jako sześcian 2×2×2',
    scene=dict(
        xaxis=dict(
            title='Gender',
            tickmode='array',
            tickvals=[0.5, 1.5],
            ticktext=['F', 'M'],
            range=[0, 2]
        ),
        yaxis=dict(
            title='City',
            tickmode='array',
            tickvals=[0.5, 1.5],
            ticktext=['Krakow', 'Warsaw'],
            range=[0, 2]
        ),
        zaxis=dict(
            title='Passed',
            tickmode='array',
            tickvals=[0.5, 1.5],
            ticktext=['No', 'Yes'],
            range=[0, 2]
        ),
        aspectmode='cube'
    ),
    width=900,
    height=750
)

fig.show()
````

## A.2 Rozwiązanie mini zadań

````python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# ZBIÓR 1: TIPS
# =========================

df_tips = sns.load_dataset("tips")

print("TIPS shape:", df_tips.shape)
display(df_tips.head())


# Zadanie 1
# Policz liczbę obserwacji w każdej grupie

print("\nZadanie 1 - liczba obserwacji w grupach (day):")
print(df_tips.groupby("day").size())

print("\nZadanie 1 - liczba obserwacji w grupach (sex):")
print(df_tips.groupby("sex").size())

print("\nZadanie 1 - liczba obserwacji w grupach (smoker):")
print(df_tips.groupby("smoker").size())


# Zadanie 2
# Policz średnią wybranej kolumny liczbowej dla każdej grupy

print("\nZadanie 2 - średni total_bill w grupach day:")
print(df_tips.groupby("day")["total_bill"].mean())

print("\nZadanie 2 - średni tip w grupach sex:")
print(df_tips.groupby("sex")["tip"].mean())


# Zadanie 3
# Porównaj średnią i medianę w grupach

print("\nZadanie 3 - średnia i mediana total_bill w grupach day:")
print(df_tips.groupby("day")["total_bill"].agg(["mean", "median"]))

print("\nZadanie 3 - średnia i mediana tip w grupach smoker:")
print(df_tips.groupby("smoker")["tip"].agg(["mean", "median"]))


# Zadanie 4
# Wybierz tylko część danych spełniającą warunek i policz statystyki ponownie

df_tips_smokers = df_tips[df_tips["smoker"] == "Yes"]

print("\nZadanie 4 - tylko osoby palące, średni total_bill w grupach day:")
print(df_tips_smokers.groupby("day")["total_bill"].mean())

print("\nZadanie 4 - tylko osoby palące, średnia i mediana tip w grupach sex:")
print(df_tips_smokers.groupby("sex")["tip"].agg(["mean", "median"]))


# Zadanie 5
# Narysuj histogram wybranej kolumny

plt.figure(figsize=(8, 5))
df_tips["total_bill"].plot(kind="hist", bins=20)
plt.title("Histogram total_bill - tips")
plt.xlabel("total_bill")
plt.ylabel("Liczba obserwacji")
plt.grid(alpha=0.2)
plt.show()


# Zadanie 6
# Narysuj boxplot tej samej kolumny w podziale na grupy

plt.figure(figsize=(8, 5))
df_tips.boxplot(column="total_bill", by="day")
plt.title("Boxplot total_bill w grupach day - tips")
plt.suptitle("")
plt.xlabel("day")
plt.ylabel("total_bill")
plt.grid(alpha=0.2)
plt.show()



````

````python
# =========================
# ZBIÓR 2: TITANIC
# =========================

df_titanic = sns.load_dataset("titanic")

print("TITANIC shape:", df_titanic.shape)
display(df_titanic.head())


# Zadanie 1
# Policz liczbę obserwacji w każdej grupie

print("\nZadanie 1 - liczba obserwacji w grupach sex:")
print(df_titanic.groupby("sex").size())

print("\nZadanie 1 - liczba obserwacji w grupach class:")
print(df_titanic.groupby("class").size())

print("\nZadanie 1 - liczba obserwacji w grupach embarked:")
print(df_titanic.groupby("embarked").size())


# Zadanie 2
# Policz średnią wybranej kolumny liczbowej dla każdej grupy

print("\nZadanie 2 - średni wiek age w grupach sex:")
print(df_titanic.groupby("sex")["age"].mean())

print("\nZadanie 2 - średnia fare w grupach class:")
print(df_titanic.groupby("class")["fare"].mean())


# Zadanie 3
# Porównaj średnią i medianę w grupach

print("\nZadanie 3 - średnia i mediana age w grupach sex:")
print(df_titanic.groupby("sex")["age"].agg(["mean", "median"]))

print("\nZadanie 3 - średnia i mediana fare w grupach class:")
print(df_titanic.groupby("class")["fare"].agg(["mean", "median"]))


# Zadanie 4
# Wybierz tylko część danych spełniającą warunek i policz statystyki ponownie

df_titanic_survived = df_titanic[df_titanic["survived"] == 1]

print("\nZadanie 4 - tylko osoby, które przeżyły, średni age w grupach sex:")
print(df_titanic_survived.groupby("sex")["age"].mean())

print("\nZadanie 4 - tylko osoby, które przeżyły, średnia i mediana fare w grupach class:")
print(df_titanic_survived.groupby("class")["fare"].agg(["mean", "median"]))


# Zadanie 5
# Narysuj histogram wybranej kolumny

plt.figure(figsize=(8, 5))
df_titanic["age"].dropna().plot(kind="hist", bins=20)
plt.title("Histogram age - titanic")
plt.xlabel("age")
plt.ylabel("Liczba obserwacji")
plt.grid(alpha=0.2)
plt.show()


# Zadanie 6
# Narysuj boxplot tej samej kolumny w podziale na grupy

plt.figure(figsize=(8, 5))
df_titanic.boxplot(column="age", by="sex")
plt.title("Boxplot age w grupach sex - titanic")
plt.suptitle("")
plt.xlabel("sex")
plt.ylabel("age")
plt.grid(alpha=0.2)
plt.show()
````
