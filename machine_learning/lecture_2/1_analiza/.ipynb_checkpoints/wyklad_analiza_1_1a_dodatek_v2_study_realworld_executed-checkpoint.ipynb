## 3. Ufuncs i agregacje


### Co tu pokazujemy?

NumPy pozwala wykonywać operacje na całych tablicach naraz, bez pisania pętli w Pythonie.  
Takie funkcje nazywamy często **ufuncs** (`universal functions`), np. `np.abs`, `np.sqrt`, `np.log`.

Operacje te są wykonywane w zoptymalizowanym kodzie C, dlatego są znacznie szybsze niż pętle w Pythonie.

### Co warto zapamiętać?

- agregacje dają jedną liczbę podsumowującą całą tablicę, np. średnią czy medianę,
- percentyle pokazują, gdzie leżą określone progi rozkładu danych (np. 25%, 50%, 75%),
- ufuncs działają elementowo, ale na całej tablicy naraz (wektorowo).

### Interpretacja tutaj

- średnia i mediana ocen są do siebie zbliżone,
- percentyle pomagają opisać typowy zakres wyników,
- `abs(ocena - średnia)` pokazuje, kto jest blisko średniej, a kto odstaje  
  (można to też interpretować jako odległość — metryka L1 w 1D),
- `sqrt(StudyHours)` to przykład funkcji działającej na całej kolumnie naraz.

`````python
print('Średnia ocen:', grades_arr.mean())
print('Mediana ocen:', np.median(grades_arr))
print('Wartości percentyli 25/50/75:', np.percentile(grades_arr, [25, 50, 75])) # jak nazywamy te percentyle? :)

distance_from_mean = np.abs(grades_arr - grades_arr.mean())
print('\nOdległość L1 (bezwzględne odchylenie) każdej oceny od średniej (szczegóły poniżej):')
print(distance_from_mean.round(2))

print('\nPierwiastek z czasu nauki (ufunc):')
print(np.sqrt(study_arr).round(3))
`````

Bezwzględne odchylenie od średniej:
$$
|x_i - \bar{x}|
$$

można interpretować jako odległość.

W 1D:
$$
d_1(a, b) = |a - b|
$$

Czyli: **bezwzględne odchylenie = odległość L1 od średniej (w 1D)**.

## 4. Broadcasting i standaryzacja

### Co tu pokazujemy?

Broadcasting to mechanizm NumPy i pandas, dzięki któremu możemy łączyć kolumnę z jedną liczbą, np. średnią lub odchyleniem standardowym, bez ręcznego powielania wartości.

### Co robimy tutaj?

Tworzymy zmienne standaryzowane:
- `StudyHours_z`
- `Grade_z`

czyli przekształcamy dane do skali:
- średnia = 0
- odchylenie standardowe = 1

### Po co to robić?

Standaryzacja pozwala porównywać zmienne zapisane w różnych jednostkach.  
Dzięki temu łatwiej zobaczyć:
- kto jest powyżej lub poniżej średniej,
- jak bardzo dany wynik odbiega od typowego poziomu grupy.

`````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

vec = np.array([10, 20, 30])

print(arr + vec)
`````

`````python
df_broadcast = df_students.copy()
df_broadcast['StudyHours_z'] = (df_broadcast['StudyHours'] - df_broadcast['StudyHours'].mean()) / df_broadcast['StudyHours'].std(ddof=1)
df_broadcast['Grade_z'] = (df_broadcast['Grade'] - df_broadcast['Grade'].mean()) / df_broadcast['Grade'].std(ddof=1)

df_broadcast[['Name', 'StudyHours', 'StudyHours_z', 'Grade', 'Grade_z']].head(10)
`````

Odchylenie standardowe liczone jest jako:
$$
\sqrt{\frac{1}{n - \text{ddof}} \sum (x_i - \bar{x})^2}
$$

`ddof` określa, o ile zmniejszamy liczbę stopni swobody (większe `ddof` → większe odchylenie).

# 4.a)

### Broadcasting — mini ćwiczenie

Dane:

- `a` ma shape (3,1)
- `b` ma shape (3,)

Zgadnij wynik przed uruchomieniem:

1. Jaki będzie shape wyniku?
2. Jak NumPy „rozciągnie” tablice?

Wskazówka: NumPy dopasowuje wymiary od końca.

`````python
import numpy as np

a = np.array([[1],
              [2],
              [3]])   # shape (3,1)

b = np.array([10, 20, 30])  # shape (3,)

print(a + b)
print((a + b).shape)
`````

### Rozwiązanie

NumPy widzi:

- `a` → (3,1)
- `b` → (1,3)

➡️ wynik:
- shape: (3,3)

czyli każda wartość z `a` dodaje się do całego `b`.

### Broadcasting vs „mnożenie na papierze”

Weźmy dwa wektory:

- pionowy:
$$
a = \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}
$$

- poziomy:
$$
b = \begin{bmatrix} 10 & 20 & 30 \end{bmatrix}
$$

#### Broadcasting (dodawanie)

NumPy „rozciąga” je do:

$$
\begin{bmatrix}
1 & 1 & 1 \\
2 & 2 & 2 \\
3 & 3 & 3
\end{bmatrix}
+
\begin{bmatrix}
10 & 20 & 30 \\
10 & 20 & 30 \\
10 & 20 & 30
\end{bmatrix}
=
\begin{bmatrix}
11 & 21 & 31 \\
12 & 22 & 32 \\
13 & 23 & 33
\end{bmatrix}
$$

-- To jest **broadcasting (operacja elementowa)**

---

#### „Mnożenie na papierze” (outer product)

Jeśli zamiast dodawania robimy mnożenie:

$$
a \cdot b =
\begin{bmatrix}
1 \cdot 10 & 1 \cdot 20 & 1 \cdot 30 \\
2 \cdot 10 & 2 \cdot 20 & 2 \cdot 30 \\
3 \cdot 10 & 3 \cdot 20 & 3 \cdot 30
\end{bmatrix}
$$

-- To jest tzw. **outer product**

---

### Wniosek

- broadcasting = dopasowanie wymiarów + operacja elementowa  
- outer product = specjalny przypadek mnożenia (bez sumowania)  
- mnożenie macierzy (klasyczne) = jeszcze coś innego (sumy iloczynów)

**Broadcasting nie kopiuje danych w pamięci — to tylko interpretacja wymiarów.**

`````python
import numpy as np

a = np.array([[1],
              [2],
              [3]])   # (3,1)

b = np.array([10, 20, 30])  # (3,)

# broadcasting (dodawanie)
print(a + b)

# outer product (mnożenie elementowe z broadcastem)
print(a * b)

# klasyczne mnożenie macierzy
print(a @ b.reshape(1, -1))  # shape (3,3)
`````

W Pythonie i NumPy przypisanie **nie tworzy kopii danych**.

```python

````

oznacza:

* `a` i `b` wskazują na **ten sam obiekt w pamięci**

Aby stworzyć niezależną kopię:

```python
b = a.copy()
```

* `b` to nowa tablica — zmiany nie wpływają na `a`

## 5. Maski boolowskie

### Co tu pokazujemy?

Maska boolowska to tablica wartości `True/False`, która mówi, które wiersze spełniają warunek.

### Dlaczego to jest ważne?

Maska nie służy tylko do filtrowania. Możemy ją też:
- liczyć (`sum()`),
- uśredniać (`mean()`),
- łączyć z innymi maskami przez `&`, `|`, `~`.

### Interpretacja tutaj

- kto ma ocenę co najmniej 60,
- kto uczył się co najmniej 12 godzin,
- kto spełnia oba warunki naraz.

## 6. Sortowanie i top-N

Praktyczny odpowiednik `argsort`, `sort_values`, `nlargest`.

### Co tu pokazujemy?

Sortowanie pozwala szybko znaleźć największe i najmniejsze wartości oraz ustalić kolejność rekordów.

### Co warto zapamiętać?
- `argsort` zwraca indeksy, które posortowałyby dane — czyli mówi, w jakiej kolejności ustawić elementy, aby były uporządkowane rosnąco.
- `sort_values()` porządkuje dane według wybranej kolumny,
- `nlargest()` i `nsmallest()` są wygodne, gdy chcemy tylko kilka najlepszych lub najsłabszych wyników.

### Interpretacja tutaj

To prosty sposób, żeby zobaczyć:
- kto ma najwyższe oceny,
- kto uczył się najdłużej,
- czy najwyższe oceny zawsze idą w parze z największą liczbą godzin.

`````python
grade_order = np.argsort(df_students['Grade'].to_numpy()) #malejaco [::-1]
print('Indeksy po sortowaniu rosnącym:', grade_order)

print('\nTop 5 ocen:')
display(df_students.nlargest(5, 'Grade')[['Name', 'StudyHours', 'Grade']])

print('\nNajmniej czasu nauki:')
display(df_students.nsmallest(5, 'StudyHours')[['Name', 'StudyHours', 'Grade']])
`````

## 8. Braki danych: `isnull`, `fillna`, `dropna`
Tutaj świadomie psujemy mały fragment danych, żeby pokazać pełny workflow naprawy.

### Co tu pokazujemy?

W realnych danych często pojawiają się braki (`NaN`).  
Musimy zdecydować, co z nimi zrobić:
- usunąć wiersze,
- uzupełnić brak,
- albo zostawić świadomie.

- `isnull()` – wskazuje, gdzie w danych znajdują się brakujące wartości (NaN)  
- `fillna()` – uzupełnia brakujące wartości podaną wartością  
- `dropna()` – usuwa wiersze lub kolumny zawierające brakujące dane  

### Co warto zapamiętać?

Braki danych to nie tylko „problem techniczny”.  
To część analizy, bo wybór metody wpływa na wynik.

## 9. `concat` i `merge`
To jest mały krok od jednej tabeli do łączenia kilku źródeł danych.

### Co tu pokazujemy?

Czasem dane są rozbite na kilka tabel i trzeba je połączyć.

- `concat()` skleja dane „jedne pod drugimi” albo „obok siebie”,
- `merge()` łączy tabele po wspólnym kluczu, np. po nazwie lub identyfikatorze.

### Dlaczego to jest ważne?

W praktyce analiza danych bardzo często zaczyna się właśnie od poprawnego połączenia źródeł danych.

### merge — jak join w SQL

`merge()` działa podobnie do `JOIN` w SQL — łączy dwie tabele na podstawie wspólnego klucza.

Najważniejsze parametry:

- `on` — kolumna (lub kolumny), po której łączymy dane  
  (np. `on='Name'`)

- `how` — sposób łączenia:
  - `'inner'` — tylko wspólne rekordy (przecięcie)
  - `'left'` — wszystkie z lewej tabeli + dopasowania z prawej
  - `'right'` — wszystkie z prawej tabeli + dopasowania z lewej
  - `'outer'` — wszystkie rekordy z obu tabel (suma)

### Intuicja

`merge` dopasowuje wiersze z dwóch tabel tam, gdzie wartości w kolumnie `on` są takie same.

`````python
df_meta = pd.DataFrame({
    'Name': names,
    'City': ['Kraków', 'Warszawa', 'Kraków', 'Wrocław', 'Poznań', 'Warszawa',
             'Gdańsk', 'Kraków', 'Łódź', 'Poznań', 'Warszawa', 'Kraków',
             'Wrocław', 'Gdańsk', 'Poznań', 'Łódź', 'Kraków', 'Warszawa',
             'Gdańsk', 'Wrocław', 'Poznań', 'Kraków'],
    'Track': ['A', 'B', 'A', 'A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'A',
              'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C', 'A', 'B']
})

extra_flag = pd.Series(df_students['StudyHours'] >= 12, name='LongStudy')
print(extra_flag)
print('concat (doklejenie kolumny):')
display(pd.concat([df_students, extra_flag], axis=1).head())

print('merge (łączenie po Name):')
display(pd.merge(df_students, df_meta, on='Name', how='left').head(10))
`````

## 10. `groupby().agg()`
To jest najważniejsze rozszerzenie części o `groupby`: kilka statystyk naraz, czytelnie i w jednej tabeli.

### Co tu pokazujemy?

`groupby()` pozwala dzielić dane na grupy, a potem liczyć statystyki osobno dla każdej grupy.

### Co daje `agg()`?

Możemy policzyć kilka miar naraz, np.:
- średnią,
- medianę,
- liczność,
- minimum i maksimum.

### Interpretacja tutaj

To pierwszy krok od patrzenia na pojedyncze osoby do patrzenia na całe grupy.

Np. przydaje się dla czegoś w stylu podusmowujących statystyk.

`````python

group_summary = (
    df_students.groupby('Passed')
    .agg(
        n=('Name', 'size'),
        mean_grade=('Grade', 'mean'),
        median_grade=('Grade', 'median'),
        mean_hours=('StudyHours', 'mean'),
        max_grade=('Grade', 'max')
    )
    .reset_index()
)

group_summary
`````

`reset_index()` zamienia indeks z powrotem w zwykłą kolumnę i tworzy nowy prosty indeks 0, 1, 2, ...

## 11. `transform()`
`agg()` daje jedną wartość na grupę, a `transform()` zwraca wynik w rozmiarze oryginalnej tabeli.

### Co tu pokazujemy?

`transform()` liczy statystykę w obrębie grup, ale zwraca wynik w oryginalnym rozmiarze tabeli.

### Po co to jest przydatne?

Dzięki temu możemy porównać pojedynczego studenta nie do całej grupy, ale do jego własnej podgrupy, np. średniej swojej kategorii.

`````python

df_transform = df_students.copy()
df_transform['Grade_vs_pass_group_mean'] = (
    df_transform['Grade'] - df_transform.groupby('Passed')['Grade'].transform('mean')
)

df_transform[['Name', 'Passed', 'Grade', 'Grade_vs_pass_group_mean']].head(10)
`````

`transform()` nie zmniejsza tabeli, tylko zwraca wartość dla każdego wiersza, liczoną w obrębie jego grupy.

## 12. `cut` i `qcut`
To jest bardzo wygodne przed `groupby` i `pivot_table`: sami budujemy grupy przedziałowe.

### Co tu pokazujemy?

Czasem zamiast dokładnych liczb wygodniej pracować na przedziałach.

- `cut()` dzieli dane według ustalonych granic,
- `qcut()` dzieli dane na grupy o podobnej liczebności.

### Po co to robić?

To pomaga budować kategorie, np.:
- mało / średnio / dużo godzin nauki,
- niski / średni / wysoki wynik.

`````python
df_bands = df_students.copy()
df_bands['StudyBand'] = pd.cut(
    df_bands['StudyHours'],
    bins=[0, 8, 10, 12, 20],
    labels=['<=8', '8-10', '10-12', '12+']
)

df_bands['GradeQuartile'] = pd.qcut(
    df_bands['Grade'],
    q=4,
    duplicates='drop'
)

display(df_bands[['Name', 'StudyHours', 'StudyBand', 'Grade', 'GradeQuartile']].head(10))
display(df_bands['StudyBand'].value_counts().sort_index())
`````

`q` w `qcut` określa, na ile równolicznych grup dzielimy dane (np. `q=4` → kwartyle).

`````python
df_test = df_students.copy()
df_test["StudyBand"] = pd.cut(df_test['StudyHours'],bins = [0,8,12,20], labels=['<=8','<=12','<=20'])

df_test['GradeQuartile'] = pd.qcut(df_test['Grade'], q = 4, duplicates='drop')

display(df_test.head())
display(df_test['StudyBand'].value_counts().sort_index())
`````

## 14. Stringi w Pandas
To będzie bardzo przydatne później przy analizach: czyszczenie napisów i standaryzacja odpowiedzi.


W praktyce dane tekstowe rzadko są od razu „czyste”.  
Nazwy, kategorie i odpowiedzi tekstowe mogą mieć:
- zbędne spacje,
- różną wielkość liter,
- literówki lub różne warianty zapisu,
- dodatkowe fragmenty, które trzeba wydzielić.

Pandas udostępnia do tego accessor `str`, dzięki któremu możemy wykonywać operacje tekstowe na całej kolumnie naraz, bez pisania klasycznej pętli.

`````python

df_names = df_students[['Name']].copy()
df_names['NameMessy'] = ['  ' + n.lower() + '  ' if i % 2 == 0 else n.upper() for i, n in enumerate(df_names['Name'])]

df_names['NameClean'] = (
    df_names['NameMessy']
    .str.strip()
    .str.title()
)

df_names['Contains_A'] = df_names['NameClean'].str.contains('a', case=False)
df_names.head(10)
`````

## Operacje na stringach w Pandas

Pandas udostępnia wygodny interfejs `.str`, który pozwala wykonywać operacje na całej kolumnie tekstowej jednocześnie (wektorowo).

Najważniejsze operacje:

### 1. Czyszczenie tekstu
- `str.strip()` – usuwa spacje z początku i końca
- `str.lstrip()` – usuwa spacje z lewej
- `str.rstrip()` – usuwa spacje z prawej

### 2. Zmiana wielkości liter
- `str.lower()` – małe litery
- `str.upper()` – wielkie litery
- `str.title()` – Każde Słowo Wielką Literą

### 3. Informacje o tekście
- `str.len()` – długość napisu

### 4. Wyszukiwanie
- `str.contains('tekst')` – czy zawiera fragment
- `str.startswith('A')` – czy zaczyna się od
- `str.endswith('Z')` – czy kończy się na

### 5. Zamiana tekstu
- `str.replace('a', 'b')` – zamiana znaków/fragmentów

### 6. Dzielenie i wyciąganie części
- `str.split(' ')` – dzieli tekst na listę
- `str.split('@').str[0]` – dostęp do elementu po podziale
- `str.extract(r'pattern')` – wyciąganie wzorca (regex)

### 7. Sprawdzanie typu znaków
- `str.isalpha()` – tylko litery
- `str.isdigit()` – tylko cyfry

### 8. Obsługa braków danych
- `fillna('')` – uzupełnia NaN przed operacjami tekstowymi

---

Najczęściej używany schemat:

df['col'].str.<operacja>()

`````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['  alice  ', 'BOB', '  Charlie', 'david  ', None],
    'Email': ['alice@gmail.com', 'bob@yahoo.com', 'charlie@wp.pl', 'david@gmail.com', None]
})
display(df)
`````

### Mini ćwiczenie

1. Usuń spacje z kolumny Name

2. Zamień wszystkie imiona na format "Alice"

3. Oblicz długość każdego imienia

4. Sprawdź, które osoby mają literę "a" w imieniu (case insensitive)

5. Sprawdź, które maile są z Gmaila

6. Wyciągnij nazwę domeny z Email (np. gmail.com)

7. Wyciągnij tylko nazwę użytkownika z Email (przed @)

8. Zamień "gmail.com" na "google.com"

9. Upewnij się, że operacje działają mimo NaN (hint: fillna)

10. Stwórz nową kolumnę:
 "Alice (gmail.com)"

`````python
# Ćwiczenie 1 i 2
# 1. Usuń spacje z kolumny Name
# 2. Zamień wszystkie imiona na format "Alice"
df['Name_clean'] = df['Name'].fillna('').str.strip().str.title()

# Ćwiczenie 3
# 3. Oblicz długość każdego imienia
df['Name_len'] = df['Name_clean'].str.len()

# Ćwiczenie 4
# 4. Sprawdź, które osoby mają literę "a" w imieniu (case insensitive)
df['has_a'] = df['Name_clean'].str.contains('a', case=False)

# Ćwiczenie 5
# 5. Sprawdź, które maile są z Gmaila
df['is_gmail'] = df['Email'].fillna('').str.contains('gmail.com')

# Ćwiczenie 6
# 6. Wyciągnij nazwę domeny z Email (np. gmail.com)
df['domain'] = df['Email'].fillna('').str.split('@').str[1]

# Ćwiczenie 7
# 7. Wyciągnij tylko nazwę użytkownika z Email (przed @)
df['user'] = df['Email'].fillna('').str.split('@').str[0]

# Ćwiczenie 8
# 8. Zamień "gmail.com" na "google.com"
df['Email_fixed'] = df['Email'].fillna('').str.replace('gmail.com', 'google.com', regex=False)

# Ćwiczenie 9
# 9. Upewnij się, że operacje działają mimo NaN (hint: fillna)
# fillna('') zostało użyte w powyższych operacjach tekstowych

# Ćwiczenie 10
# 10. Stwórz nową kolumnę: "Alice (gmail.com)"
df['Summary'] = df['Name_clean'] + ' (' + df['domain'].fillna('') + ')'

display(df)
`````

`````python
# Ćw. 1-2: czyszczenie i formatowanie imion
df['Name_clean'] = df['Name'].fillna('').str.strip().str.title()

# Ćw. 3: długość imienia
df['Name_len'] = df['Name_clean'].str.len()

# Ćw. 4: litera "a" w imieniu
df['has_a'] = df['Name_clean'].str.contains('a', case=False)

# Ćw. 5: czy email jest z Gmaila
df['is_gmail'] = df['Email'].fillna('').str.contains('gmail.com')

# Ćw. 6: domena email
df['domain'] = df['Email'].fillna('').str.split('@').str[1]

# Ćw. 7: użytkownik email
df['user'] = df['Email'].fillna('').str.split('@').str[0]

# Ćw. 8: zamiana gmail.com na google.com
df['Email_fixed'] = df['Email'].fillna('').str.replace('gmail.com', 'google.com', regex=False)

# Ćw. 9: obsługa NaN przez fillna('')
# zastosowano w operacjach powyżej

# Ćw. 10: podsumowanie tekstowe
df['Summary'] = df['Name_clean'] + ' (' + df['domain'].fillna('') + ')'

display(df)
`````

## 15. Histogram, boxplot i gęstość + PDF
Krótka teoria: histogram pokazuje dane dyskretne w koszykach, a PDF to gładkie przybliżenie rozkładu ciągłego. Pole pod PDF ma wynosić 1.

Argumenty nazwane (keyword arguments) pozwalają przekazać wartości do funkcji,
jednocześnie nadpisując wartości domyślne.

Przykład:
show_distribution(data, title='Nowy tytuł')

Jeśli nie podamy argumentu, użyta zostanie wartość domyślna z definicji funkcji.

### Co tu pokazujemy?

Histogram pokazuje rozkład danych dyskretnych z próby.  
PDF to gładka funkcja opisująca rozkład teoretyczny zmiennej ciągłej.

### Najważniejsza intuicja

- histogram opisuje nasze dane,
- PDF opisuje model rozkładu.

Porównanie ich pozwala ocenić, czy dane są zbliżone do prostego modelu statystycznego, np. rozkładu normalnego.

`````python

show_distribution(df_students['Grade'], title='Oceny: histogram + boxplot')
show_density(df_students['Grade'], title='Oceny: gęstość')

mu = df_students['Grade'].mean()
sigma = df_students['Grade'].std(ddof=1)
x = np.linspace(df_students['Grade'].min() - 10, df_students['Grade'].max() + 10, 400)
pdf = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(10, 4))
plt.hist(df_students['Grade'], bins=8, density=True, alpha=0.6, label='Histogram (density=True)')
plt.plot(x, pdf, color='crimson', linewidth=2, label='PDF N(mean, std)')
plt.title('Histogram i przykładowa krzywa PDF')
plt.xlabel('Grade')
plt.ylabel('Gęstość')
plt.legend()
plt.show()

area = np.trapz(pdf, x)
within_one_sigma = ((df_students['Grade'] >= mu - sigma) & (df_students['Grade'] <= mu + sigma)).mean()
print(f'Przybliżone pole pod PDF: {area:.4f}')
print(f'Odsetek ocen w zakresie mean ± 1 std: {within_one_sigma:.2%}')
`````

## 16. Error bars
To jest dobra mini-wstawka przy porównywaniu grup: nie tylko średnia, ale też rozrzut.

### Co tu pokazujemy?

Średnia sama w sobie nie wystarcza.  
Warto też pokazać rozrzut wyników.

### Po co są error bars?

Słupki błędów pokazują, jak bardzo wyniki w grupie się różnią.  
Dzięki temu widzimy nie tylko poziom średni, ale też stabilność i zmienność danych.

`````python

band_stats = (
    df_bands.groupby('StudyBand', observed=False)
    .agg(mean_grade=('Grade', 'mean'), std_grade=('Grade', 'std'), n=('Grade', 'size'))
    .reset_index()
)
display(band_stats.round(2))
plt.figure(figsize=(8, 4))
plt.errorbar(
    band_stats['StudyBand'].astype(str),
    band_stats['mean_grade'],
    yerr=band_stats['std_grade'],
    fmt='o-',
    capsize=5
)
plt.title('Średnia ocena ± odchylenie standardowe')
plt.xlabel('StudyBand')
plt.ylabel('Grade')
plt.show()

band_stats.round(2)
`````

## 17. Scatter, korelacja i regresja
To naturalne rozwinięcie `realworldexploration`.

### Co tu pokazujemy?

Wykres punktowy (`scatter`) pozwala zobaczyć zależność między dwiema zmiennymi.

### Interpretacja tutaj

Patrzymy, czy większa liczba godzin nauki wiąże się zwykle z wyższą oceną.  
Linia regresji nie daje „prawdy”, ale pomaga zobaczyć ogólny trend.

`````python

corr = df_students['StudyHours'].corr(df_students['Grade'])
m, b, r, p, se = stats.linregress(df_students['StudyHours'], df_students['Grade'])

x_line = np.linspace(df_students['StudyHours'].min(), df_students['StudyHours'].max(), 100)
y_line = m * x_line + b

plt.figure(figsize=(7, 4))
plt.scatter(df_students['StudyHours'], df_students['Grade'], c=df_students['Passed'].map({True: 'teal', False: 'orange'}), alpha=0.8)
plt.plot(x_line, y_line, color='crimson', linewidth=2)
plt.title(f'StudyHours vs Grade | corr={corr:.3f}')
plt.xlabel('StudyHours')
plt.ylabel('Grade')
plt.show()

print(f'm = {m:.4f}, b = {b:.4f}, r = {r:.4f}, p = {p:.4g}')
`````

## 19. `hexbin` jako opcjonalny bonus
Na małym zbiorze to nie ma sensu, więc generujemy większą próbkę syntetyczną wokół tych samych danych.

Przy większej liczbie punktów wykres scatter przestaje być czytelny (punkty się nakładają).  
Hexbin grupuje dane w sześciokąty i pokazuje ich zagęszczenie kolorem, dzięki czemu łatwiej zobaczyć strukturę i trend.

`````python

rng = np.random.default_rng(42)
base = df_students[['StudyHours', 'Grade']].to_numpy()
rep = np.repeat(base, 25, axis=0)
noise = rng.normal(loc=0.0, scale=[0.35, 3.5], size=rep.shape)
big = pd.DataFrame(rep + noise, columns=['StudyHours', 'Grade'])

plt.figure(figsize=(7, 4))
plt.hexbin(big['StudyHours'], big['Grade'], gridsize=18, cmap='viridis')
plt.colorbar(label='liczba punktów w koszyku')
plt.title('Hexbin na większej próbie syntetycznej')
plt.xlabel('StudyHours')
plt.ylabel('Grade')
plt.show()
`````
