# Mapa matematyki z poprzedniej wersji

Poniższe sekcje zostawiam jako pomocniczą mapę pojęć. Można je przejść szybko albo potraktować jako materiał do samodzielnej powtórki.

## 1. Co trzeba domknąć po wykładzie 3?

Z notatek po wykładzie najważniejszy jest powrót do trzech różnych obiektów:

| Symbol | Znaczenie | Gdzie używamy? |
|---|---|---|
| $\bar y$ | średnia targetu | baseline regresyjny, $TSS$, $R^2$ |
| $\hat y$ | predykcja wartości liczbowej | regresja, błąd, $MAE$, $RMSE$, $R^2$ |
| $\hat p$ | predykcja prawdopodobieństwa klasy 1 | regresja logistyczna, ROC/AUC, threshold |

To rozróżnienie jest bardzo ważne, bo potem wiele nieporozumień bierze się z mieszania tych trzech rzeczy.

Przykład:

$$
\bar y = \text{średnia ocena w grupie}
$$

$$
\hat y_i = \text{przewidywana ocena konkretnej osoby}
$$

$$
\hat p_i = P(\text{zdał} \mid X_i)
$$

Dopiero z $\hat p_i$ robimy twardą decyzję:

$$
\hat y_i^{class} = I(\hat p_i \ge t)
$$

czyli wybieramy próg $t$.

## 2. Minimalna matematyka dla domknięcia poprzednich tematów

### 2.1. Sumy i średnie

W regresji potrzebujemy rozumieć:

$$
\bar y = \frac{1}{n}\sum_{i=1}^{n} y_i
$$

oraz błędy:

$$
y_i - \hat y_i
$$

### 2.2. Kwadraty błędów

Dla $R^2$ potrzebne są dwie sumy kwadratów:

$$
SSE = \sum_i (y_i - \hat y_i)^2
$$

$$
TSS = \sum_i (y_i - \bar y)^2
$$

A potem:

$$
R^2 = 1 - \frac{SSE}{TSS}
$$

Intuicja:

- $TSS$ — jak bardzo dane różnią się od średniej,
- $SSE$ — jak bardzo dane różnią się od predykcji modelu,
- $R^2$ — czy model jest lepszy od zgadywania średniej.

## 3. Minimalna matematyka dla regresji logistycznej

Potrzebujemy trzech przekształceń:

### 3.1. Prawdopodobieństwo

$$
0 \le p \le 1
$$

### 3.2. Odds

$$
odds = \frac{p}{1-p}
$$

Przykład: jeśli $p=0.8$, to:

$$
odds = \frac{0.8}{0.2}=4
$$

czyli sukces ma odds $4:1$.

### 3.3. Log-odds

$$
logit(p)=\log\left(\frac{p}{1-p}\right)
$$

Regresja logistyczna mówi:

$$
logit(p_i)=b_0+b_1x_i
$$

Czyli liniowa jest nie sama wartość $p$, tylko **logarytm oddsów**.

````python
import numpy as np
import pandas as pd

p = np.array([0.1, 0.25, 0.5, 0.75, 0.9])
odds = p / (1 - p)
log_odds = np.log(odds)

pd.DataFrame({
    'p': p,
    'odds = p/(1-p)': odds,
    'log_odds': log_odds
}).round(3)
````

## 4. Minimalna matematyka dla drzew decyzyjnych

Drzewo decyzyjne potrzebuje pojęcia **nieczystości węzła**.

Dla klasyfikacji binarnej możemy użyć Gini impurity:

$$
Gini = 1 - p_0^2 - p_1^2
$$

gdzie:

- $p_0$ to udział klasy 0 w węźle,
- $p_1$ to udział klasy 1 w węźle.

Jeżeli w węźle są same jedynki albo same zera, wtedy:

$$
Gini = 0
$$

czyli węzeł jest czysty.

Jeżeli jest pół na pół:

$$
p_0 = 0.5, \quad p_1 = 0.5
$$

wtedy:

$$
Gini = 1 - 0.5^2 - 0.5^2 = 0.5
$$

czyli węzeł jest mocno nieczysty.

````python
def gini_from_counts(n0, n1):
    n = n0 + n1
    p0 = n0 / n
    p1 = n1 / n
    return 1 - p0**2 - p1**2

rows = []
for n0, n1 in [(10, 0), (8, 2), (5, 5), (2, 8), (0, 10)]:
    rows.append({
        'class_0': n0,
        'class_1': n1,
        'p_1': n1 / (n0 + n1),
        'Gini': gini_from_counts(n0, n1)
    })

pd.DataFrame(rows).round(3)
````

## 5. Minimalna matematyka dla Random Forest

Random Forest potrzebuje dwóch intuicji:

### 5.1. Bootstrap

Losujemy próbkę **z powtórzeniami**. To znaczy, że ta sama obserwacja może trafić do próbki kilka razy, a inna może nie trafić wcale.

To pozwala zbudować wiele trochę różnych drzew.

### 5.2. Uśrednianie zmniejsza wariancję

Jeśli mamy wiele niestabilnych modeli i uśredniamy ich predykcje, to często dostajemy stabilniejszy wynik.

Dla regresji intuicja jest taka:

$$
\hat f_{forest}(x)=\frac{1}{B}\sum_{b=1}^{B}\hat f_b(x)
$$

Dla klasyfikacji drzewa głosują:

$$
\hat y = \text{większość głosów drzew}
$$

Najważniejszy komunikat:

> Random Forest nie jest jednym dużym drzewem. To wiele różnych drzew, których błędy częściowo się znoszą.

````python
np.random.seed(42)
original = np.arange(10)
bootstrap_sample = np.random.choice(original, size=len(original), replace=True)

pd.DataFrame({
    'oryginalne_indeksy': original,
    'bootstrap_sample': bootstrap_sample
})
````

# Minimalna wspólna matematyka dla całej sekwencji

Poniższe pojęcia wracają w kolejnych notebookach. Warto je pokazać nie jako osobną teorię, ale jako narzędzia używane w konkretnych modelach.

## 1. Decyzja jako próg

$$
\hat y_i = I(\hat p_i \ge t)
$$

Jeden model probabilistyczny może mieć wiele różnych macierzy pomyłek, bo zmieniamy próg $t$.

## 2. Split jako pytanie

$$
x_j \le c?
$$

To jest fundament drzew. W 2D takie pytania tworzą pionowe i poziome cięcia.

## 3. Czystość węzła

$$
Gini = 1 - \sum_k p_k^2
$$

Im bardziej w węźle dominuje jedna klasa, tym mniejsza nieczystość.

## 4. Głosowanie wielu modeli

$$
\hat p(x) = \frac{1}{B}\sum_{b=1}^{B} I(T_b(x)=1)
$$

To dobra intuicja dla Random Forest.

## 5. Model addytywny w boostingu

$$
F_M(x) = F_0(x) + \eta \sum_{m=1}^{M} f_m(x)
$$

Każde kolejne małe drzewo poprawia to, czego poprzednie modele jeszcze nie umiały.

## 6. Granica i margines w SVM

$$
f(x)=w^T x + b
$$

$$
\hat y = \text{sign}(f(x))
$$

SVM nie tylko szuka granicy, ale szuka granicy z dużym marginesem.
