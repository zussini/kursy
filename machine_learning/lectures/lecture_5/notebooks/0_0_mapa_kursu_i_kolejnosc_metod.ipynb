# 0_0. Mapa kursu: od odległości do klasyfikatorów

Ten kurs prowadzi jedną spójną historię:

```text
punkty → odległości → sąsiedztwo → graf → algorytmy
```

Najważniejsza decyzja, która wraca w wielu metodach, brzmi:

> kiedy dwa punkty uznajemy za podobne albo sąsiednie?

W notacji pojawią się głównie: odległość $d(x_i,x_j)$, promień $\varepsilon$, liczba sąsiadów $k$, macierze sąsiedztwa $A$, podobieństwa $P$ i $Q$, centroidy $\mu_c$ oraz parametr miękkości marginesu $C$.


## Proponowana kolejność prowadzenia

1. `0_1_odleglosci_macierze_sasiedztwa_knn_fundament_recznie`  
   Fundament: macierz odległości, $\varepsilon$-sąsiedztwo, kNN, graf, symetryzacja `union` i `mutual`.

2. `2_4_umap_tsne_similarity_scores_recznie`  
   Most: z odległości robimy similarity scores, czyli miękkie podobieństwa.

3. `2_3_umap_recznie_pelny_algorytm` oraz `1_1_umap_interaktywny`  
   UMAP jako budowanie fuzzy grafu i dopasowanie układu w niskim wymiarze.

4. `2_2_tsne_recznie_pelny_algorytm` oraz `1_2_tsne_interaktywny_macierz_dopasowania`  
   t-SNE jako dopasowywanie macierzy podobieństw $P$ i $Q$.

5. `2_1_pca_recznie_matematycznie` oraz `1_3_pca_kontrast_liniowy`  
   PCA jako liniowy punkt odniesienia.

6. `3_1`, `3_2`, `3_3`, `3_4`  
   KNN, DBSCAN i k-means: trzy różne odpowiedzi na pytanie o lokalność, gęstość i centroidy.

7. `4_1`, `4_2`, `4_3`, `4_4`  
   LDA, QDA i SVM: metody nadzorowane, granice decyzyjne i marginesy.


## Dlaczego `0_1` jest pierwszy?

Bo prawie wszystkie kolejne notebooki używają jednej z dwóch idei:

$$
A^{(\varepsilon)}_{ij}=1[d(x_i,x_j)\leq \varepsilon]
$$

albo

$$
A^{(kNN)}_{ij}=1[x_j \in kNN(x_i)].
$$

DBSCAN używa promienia $\varepsilon$.  
KNN używa listy $k$ sąsiadów.  
UMAP startuje od grafu sąsiedztwa.  
t-SNE buduje macierz podobieństw.  
Spectral clustering i GNN również zaczynają od grafu.
