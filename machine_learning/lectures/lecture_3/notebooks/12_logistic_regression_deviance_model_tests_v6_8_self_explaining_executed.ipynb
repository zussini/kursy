# 12. Regresja logistyczna: porównywanie modeli, deviance, pseudo-$R^2$ i testy współczynników

Ten notebook rozwija część o likelihoodach. W regresji logistycznej najczęściej porównujemy modele przez log-likelihood i deviance, a nie przez $SSE$.

Najważniejsze obiekty:

$$
\ell(\beta)=\sum_i \left[y_i\eta_i - \log(1+e^{\eta_i})\right]
$$

$$
Deviance = -2(\ell_{model}-\ell_{saturated})
$$

Dla indywidualnych obserwacji Bernoulliego model nasycony może mieć $\ell_{saturated}=0$, więc:

$$
Deviance = -2\ell_{model}
$$

Porównanie modeli zagnieżdżonych:

$$
G^2 = Deviance_{reduced}-Deviance_{full}=2(\ell_{full}-\ell_{reduced})
$$

## Jak prowadzić ten notebook?

Ten notebook najlepiej traktować jako **statystyczne domknięcie regresji logistycznej**. W poprzednich częściach student powinien już znać:

- `10` — odds, log-odds i odds ratio,
- `11` — regresję logistyczną jako model:

$$
logit(p)=X\beta
$$

Tutaj dokładamy pytanie:

> Skoro mamy już model logistyczny, to jak porównać kilka modeli i jak sprawdzić, czy dodatkowa zmienna naprawdę coś wnosi?

Główna historia:

1. model zerowy przewiduje stałe prawdopodobieństwo,
2. modele pełniejsze dodają predyktory,
3. porównujemy je przez log-likelihood i deviance,
4. pojedyncze współczynniki interpretujemy przez Wald test i $e^\beta$,
5. separacja pokazuje, po co potrzebna bywa regularizacja.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2, norm

pd.set_option('display.precision', 6)
```

```python
def sigmoid(z):
    z = np.asarray(z, dtype=float)
    z = np.clip(z, -35, 35)
    return 1 / (1 + np.exp(-z))


def loglik_logistic(beta, X, y):
    eta = X @ beta
    return float(np.sum(y * eta - np.logaddexp(0, eta)))


def fit_logistic_irls(X, y, l2=0.0, max_iter=150, tol=1e-10):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)
    n, k = X.shape
    beta = np.zeros(k)
    penalty = np.eye(k)
    penalty[0, 0] = 0.0
    converged = False
    for iteration in range(max_iter):
        eta = X @ beta
        p = sigmoid(eta)
        W = p * (1 - p)
        gradient = X.T @ (y - p) - l2 * (penalty @ beta)
        information = (X.T * W) @ X + l2 * penalty
        step = np.linalg.pinv(information) @ gradient
        beta_new = beta + step
        if np.max(np.abs(step)) < tol:
            beta = beta_new
            converged = True
            break
        beta = beta_new
    eta = X @ beta
    p = sigmoid(eta)
    W = p * (1 - p)
    information = (X.T * W) @ X + l2 * penalty
    cov_beta = np.linalg.pinv(information)
    se_beta = np.sqrt(np.diag(cov_beta))
    return {
        'beta': beta,
        'se': se_beta,
        'cov': cov_beta,
        'll': loglik_logistic(beta, X, y),
        'p_hat': p,
        'converged': converged,
        'iterations': iteration + 1,
    }


def fit_named_model(df, y_col, columns, l2=0.0):
    X_parts = [np.ones(len(df))]
    names = ['intercept']
    for col in columns:
        X_parts.append(df[col].to_numpy())
        names.append(col)
    X = np.column_stack(X_parts)
    y = df[y_col].to_numpy()
    fit = fit_logistic_irls(X, y, l2=l2)
    return X, y, names, fit
```

## Mini-przypomnienie: czym różni się ten notebook od ROC/AUC?

W ROC/AUC interesuje nas ranking score'ów i progi decyzji.

Tutaj interesuje nas inna rzecz:

> czy model z dodatkowymi kolumnami w $X$ ma większy log-likelihood niż model prostszy?

Czyli:

$$
\ell_{full} > \ell_{reduced}
$$

Jeżeli poprawa jest duża, to różnica deviance może być istotna statystycznie.

## 1. Dane: ekspresja genu, masa myszy, mutacja i choroba

To mały syntetyczny przykład do pokazania idei. Mamy trzy potencjalne predyktory:

- `expression`: ekspresja genu,
- `weight`: masa myszy,
- `mutated`: informacja 0/1, czy gen jest zmutowany.

Cel: przewidzieć `disease`.

```python
rng = np.random.default_rng(7)
n = 36
weight = rng.normal(22, 3.0, n).round(1)
mutated = rng.binomial(1, 0.42, n)
expression = (1.2 + 0.08*weight + 0.55*mutated + rng.normal(0, 0.35, n)).round(2)
eta_true = -5.6 + 1.25*expression + 0.75*mutated + 0.015*weight
p_true = sigmoid(eta_true)
disease = rng.binomial(1, p_true)

# Bezpiecznik: gdyby losowanie dało zbyt mało jednej klasy, deterministycznie korygujemy kilka wartości.
if disease.sum() < 6:
    disease[np.argsort(p_true)[-6:]] = 1
if (1 - disease).sum() < 6:
    disease[np.argsort(p_true)[:6]] = 0

clinical_df = pd.DataFrame({
    'weight': weight,
    'mutated': mutated,
    'expression': expression,
    'disease': disease,
})
clinical_df.head(10)
```

```python
clinical_df['disease'].value_counts().rename('count')
```

### Jak czytać te dane?

`disease` jest zmienną binarną:

- `1` — choroba / przypadek pozytywny,
- `0` — brak choroby / przypadek negatywny.

Predyktory:

- `expression` — ciągła cecha biologiczna,
- `mutated` — cecha binarna $0/1$,
- `weight` — cecha ciągła.

To jest mały przykład dydaktyczny, więc nie chodzi o biologię, tylko o mechanikę porównywania modeli.

## 2. Tabela modeli: null, jeden predyktor, kilka predyktorów

Dla każdego modelu policzymy:

- liczbę parametrów $k$,
- log-likelihood $\ell$,
- deviance $D=-2\ell$,
- AIC:

$$
AIC = -2\ell + 2k
$$

- pseudo-$R^2$ McFaddena względem modelu zerowego:

$$
R^2_{McFadden}=1-\frac{\ell_{model}}{\ell_{null}}
$$

```python
models = {
    'null': [],
    'expression': ['expression'],
    'expression + mutated': ['expression', 'mutated'],
    'expression + mutated + weight': ['expression', 'mutated', 'weight'],
}

fits = {}
rows = []
ll_null = None
for name, cols in models.items():
    X_m, y_m, coef_names, fit_m = fit_named_model(clinical_df, 'disease', cols)
    fits[name] = {'X': X_m, 'y': y_m, 'coef_names': coef_names, 'fit': fit_m, 'columns': cols}
    if name == 'null':
        ll_null = fit_m['ll']
    k = X_m.shape[1]
    rows.append({
        'model': name,
        'k': k,
        'log_likelihood': fit_m['ll'],
        'deviance': -2*fit_m['ll'],
        'AIC': -2*fit_m['ll'] + 2*k,
        'McFadden_R2': 1 - fit_m['ll']/ll_null if ll_null is not None else 0.0,
        'converged': fit_m['converged'],
        'iterations': fit_m['iterations'],
    })

model_table = pd.DataFrame(rows)
model_table
```

### Jak czytać tabelę modeli?

Najważniejsze kolumny:

- `k` — liczba parametrów w modelu,
- `log_likelihood` — im większy, tym lepiej,
- `deviance` — $D=-2\ell$, im mniejsza, tym lepiej,
- `AIC` — kara za liczbę parametrów, im mniejszy, tym lepiej,
- `McFadden_R2` — poprawa względem modelu zerowego.

Dobry komentarz na zajęciach:

> Model pełniejszy prawie zawsze dopasuje trening lepiej. Dlatego patrzymy nie tylko na log-likelihood, ale też na AIC albo testujemy, czy poprawa jest wystarczająco duża.

```python
# Mała tabela pomocnicza: ile poprawia się model względem poprzedniego?
# To nie jest formalny test, tylko czytelny podgląd postępu.

model_progress = model_table.copy()
model_progress['delta_deviance_vs_previous'] = model_progress['deviance'].shift(1) - model_progress['deviance']
model_progress['delta_AIC_vs_previous'] = model_progress['AIC'].shift(1) - model_progress['AIC']
model_progress[['model', 'k', 'deviance', 'delta_deviance_vs_previous', 'AIC', 'delta_AIC_vs_previous', 'McFadden_R2']]
```

## 3. Likelihood-ratio test dla modeli zagnieżdżonych

Jeśli model pełny ma więcej kolumn i zawiera model prostszy jako przypadek szczególny, porównujemy:

$$
G^2 = 2(\ell_{full}-\ell_{reduced})
$$

Stopnie swobody:

$$
df = k_{full}-k_{reduced}
$$

Przykład: czy `mutated` dodaje informację po uwzględnieniu `expression`?

```python
def likelihood_ratio_test(reduced_name, full_name):
    red = fits[reduced_name]
    full = fits[full_name]
    ll_red = red['fit']['ll']
    ll_full = full['fit']['ll']
    k_red = red['X'].shape[1]
    k_full = full['X'].shape[1]
    G2 = 2 * (ll_full - ll_red)
    df = k_full - k_red
    p = chi2.sf(G2, df)
    return pd.Series({
        'reduced': reduced_name,
        'full': full_name,
        'G^2': G2,
        'df': df,
        'p_value': p,
    })

pd.DataFrame([
    likelihood_ratio_test('null', 'expression'),
    likelihood_ratio_test('expression', 'expression + mutated'),
    likelihood_ratio_test('expression + mutated', 'expression + mutated + weight'),
])
```

### LRT, AIC i Wald — kiedy czego używamy?

**Likelihood-ratio test** porównuje dwa modele zagnieżdżone, np.:

$$
\text{expression} \quad \subset \quad \text{expression + mutated}
$$

Wtedy:

$$
G^2=2(\ell_{full}-\ell_{reduced})
$$

**AIC** można traktować jako szybkie kryterium wyboru modelu, które karze złożoność:

$$
AIC=-2\ell+2k
$$

**Wald** testuje pojedynczy współczynnik:

$$
z=\frac{\hat\beta}{SE(\hat\beta)}
$$

Na zajęciach dobrze powiedzieć:

> LRT pyta, czy cały dodany blok zmiennych poprawia model. Wald pyta, czy konkretny współczynnik jest daleko od zera.

## 4. Testy Walda dla współczynników

W regresji logistycznej przybliżony test Walda dla współczynnika:

$$
z_j = \frac{\hat\beta_j}{SE(\hat\beta_j)}
$$

Jeśli $H_0:\beta_j=0$, to $z_j$ porównujemy z rozkładem normalnym standardowym.

Warto też patrzeć na $\exp(\beta_j)$, bo to jest mnożnik odds.

```python
best_name = 'expression + mutated + weight'
best = fits[best_name]
fit_best = best['fit']
coef_table = pd.DataFrame({
    'coef': fit_best['beta'],
    'std_error': fit_best['se'],
}, index=best['coef_names'])
coef_table['z_Wald'] = coef_table['coef'] / coef_table['std_error']
coef_table['p_Wald'] = 2 * norm.sf(np.abs(coef_table['z_Wald']))
coef_table['exp(coef)'] = np.exp(coef_table['coef'])
coef_table
```

### Jak interpretować `exp(coef)`?

W regresji logistycznej współczynnik $\beta$ działa na skali log-odds.

Po odlogarytmowaniu:

$$
e^\beta
$$

dostajemy mnożnik odds.

Przykład:

- jeśli $e^\beta=2$, odds rosną 2 razy,
- jeśli $e^\beta=0.5$, odds spadają o połowę,
- jeśli $e^\beta=1$, zmienna nie zmienia odds.

Uwaga: to nie jest bezpośrednia zmiana prawdopodobieństwa.

```python
# Krótkie zdaniowe interpretacje współczynników.
# Dla cech ciągłych interpretacja dotyczy wzrostu cechy o 1 jednostkę.

interpretation = coef_table.copy()
interpretation['odds_interpretation'] = interpretation['exp(coef)'].apply(
    lambda v: f"odds x {v:.2f}"
)
interpretation[['coef', 'exp(coef)', 'odds_interpretation', 'p_Wald']]
```

## 5. Model nasycony i deviance residuals — intuicja

Dla pojedynczej obserwacji binarnej model nasycony przewidziałby dokładnie $p_i=1$ dla $y_i=1$ i $p_i=0$ dla $y_i=0$. Wtedy jego log-likelihood jest równy $0$.

Dlatego residual deviance modelu Bernoulliego to:

$$
D = -2\ell_{model}
$$

Deviance residual dla obserwacji może być zapisany jako:

$$
r_i = sign(y_i-\hat p_i)\sqrt{2\left[y_i\log\left(\frac{y_i}{\hat p_i}\right)+(1-y_i)\log\left(\frac{1-y_i}{1-\hat p_i}\right)\right]}
$$

Dla $y_i=0$ lub $y_i=1$ używamy uproszczeń z granicami, bo $0\log(0)=0$.

```python
def deviance_residuals_binary(y, p):
    y = np.asarray(y)
    p = np.clip(np.asarray(p), 1e-12, 1-1e-12)
    contrib = np.where(y == 1, 2*np.log(1/p), 2*np.log(1/(1-p)))
    return np.sign(y - p) * np.sqrt(contrib)

resid_df = clinical_df.copy()
resid_df['p_hat_best'] = fit_best['p_hat']
resid_df['deviance_residual'] = deviance_residuals_binary(best['y'], fit_best['p_hat'])
resid_df.sort_values('deviance_residual', key=np.abs, ascending=False).head(8)
```

### Jak czytać deviance residuals?

Duża wartość bezwzględna reszty oznacza obserwację, która jest „zaskakująca” dla modelu.

- $y=1$, a model daje małe $p$ — duża dodatnia reszta,
- $y=0$, a model daje duże $p$ — duża ujemna reszta.

To nie jest jeszcze automatyczny powód do usunięcia punktu. To sygnał:

> sprawdź, dlaczego model był tak zaskoczony.

```python
# Wizualizacja: które obserwacje najbardziej zaskakują model?
# Im dalej od 0, tym większy wkład obserwacji do deviance.

plot_resid = resid_df.sort_values('deviance_residual').reset_index(drop=True)

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(plot_resid.index, plot_resid['deviance_residual'])
ax.axhline(0, linestyle='--')
ax.set_xlabel('obserwacje posortowane po deviance residual')
ax.set_ylabel('deviance residual')
ax.set_title('Deviance residuals: obserwacje zaskakujące dla modelu')
ax.grid(True, alpha=0.3)
plt.show()
```

## Mini-checkpoint przed separacją

Do tego momentu mamy trzy sposoby patrzenia na model logistyczny:

1. **dopasowanie całego modelu** — log-likelihood, deviance, AIC,
2. **porównanie modeli** — likelihood-ratio test,
3. **pojedyncze współczynniki** — Wald i $e^\beta$.

Separacja jest osobnym problemem: pokazuje, co się dzieje, gdy dane są „zbyt idealnie” rozdzielone.

## 6. Separacja i po co regularizacja

Jeśli klasy da się idealnie rozdzielić linią, zwykły MLE w regresji logistycznej może próbować wysłać współczynniki do nieskończoności. To nie jest błąd rachunkowy; to własność modelu.

Regularizacja L2 stabilizuje problem:

$$
NLL_{penalized}(\beta)=NLL(\beta)+\frac{\lambda}{2}\sum_{j>0}\beta_j^2
$$

Nie karzemy zwykle interceptu.

```python
sep_df = pd.DataFrame({
    'x': [-3, -2, -1, 1, 2, 3],
    'y': [0, 0, 0, 1, 1, 1]
})
X_sep = np.column_stack([np.ones(len(sep_df)), sep_df['x'].to_numpy()])
y_sep = sep_df['y'].to_numpy()

rows = []
for lam in [0.0, 0.01, 0.1, 1.0, 10.0]:
    fit_sep = fit_logistic_irls(X_sep, y_sep, l2=lam, max_iter=80)
    rows.append({
        'lambda_L2': lam,
        'intercept': fit_sep['beta'][0],
        'slope': fit_sep['beta'][1],
        'log_likelihood_unpenalized': fit_sep['ll'],
        'converged': fit_sep['converged'],
        'iterations': fit_sep['iterations'],
    })

pd.DataFrame(rows)
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(sep_df['x'], sep_df['y'])
x_grid = np.linspace(-3.5, 3.5, 300)
for lam in [0.01, 0.1, 1.0, 10.0]:
    fit_sep = fit_logistic_irls(X_sep, y_sep, l2=lam, max_iter=80)
    p_grid = sigmoid(fit_sep['beta'][0] + fit_sep['beta'][1]*x_grid)
    ax.plot(x_grid, p_grid, label=f'lambda={lam}')
ax.set_xlabel('x')
ax.set_ylabel('P(y=1)')
ax.set_title('Separacja: L2 trzyma współczynniki pod kontrolą')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

### Co ma zapamiętać student z separacji?

Jeśli klasy można idealnie rozdzielić, model logistyczny może próbować zrobić bardzo stromą sigmoidę. Wtedy współczynniki rosną, a klasyczny MLE bywa niestabilny.

Regularizacja L2 dodaje karę:

$$
-\ell(\beta)+\lambda\sum_j \beta_j^2
$$

Dzięki temu model nie musi wysyłać współczynników do nieskończoności.

To jest most do notebooka `14_regularization_ridge_lasso_elasticnet_v6`.

## 7. Mini-ćwiczenia

### Ćwiczenie A

Jeśli $\ell_{null}=-20$ i $\ell_{full}=-14$, to:

$$
G^2 = 2((-14)-(-20))=12
$$

Jeśli dodaliśmy jeden parametr, porównujemy z $\chi^2_1$.

### Ćwiczenie B

Jeśli $\beta=0.4$, to:

$$
\exp(0.4)\approx 1.49
$$

czyli wzrost predyktora o 1 jednostkę mnoży odds około $1.49$ raza.

### Ćwiczenie C

Jeśli model ma bardzo duże współczynniki i prawdopodobieństwa bliskie $0$ albo $1$, sprawdź, czy nie ma separacji albo zbyt słabej regularizacji.
