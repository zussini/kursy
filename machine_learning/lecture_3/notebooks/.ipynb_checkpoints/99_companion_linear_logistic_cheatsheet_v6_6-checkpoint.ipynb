# 99. Companion: liniowa vs logistyczna — miniściąga

Ten dodatek wyjaśnia pojęcia, które wracają w notebookach 15 i 16.

## Regresja liniowa

Model:

$$
\hat y=X\beta
$$

Funkcja celu:

$$
SSE=\sum_i(y_i-\hat y_i)^2
$$

Używamy, gdy target jest liczbą ciągłą, np. `Grade`, `final_hours`, `tip`.

## Regresja logistyczna

Model:

$$
logit(p)=\log\left(\frac{p}{1-p}\right)=X\beta
$$

Powrót do probability:

$$
p=\frac{1}{1+e^{-X\beta}}
$$

Używamy, gdy target jest binarny, np. `Pass`, `under_4h`, `survived`, `large_tip`.

## Leakage

Leakage pojawia się, gdy cecha zawiera informację, której nie mielibyśmy w momencie predykcji, albo jest przepisanym targetem.

Przykłady:

- Titanic: `alive` jako cecha do przewidywania `survived`,
- Tips: `tip` jako cecha do przewidywania `large_tip`,
- Marathon: `final_hours` jako cecha do przewidywania `under_4h`.

## Regularizacja

L2/Ridge:

$$
Loss + \lambda\sum_j\beta_j^2
$$

L1/Lasso:

$$
Loss + \lambda\sum_j|\beta_j|
$$

W `sklearn` dla logistic regression parametr `C` jest odwrotnością siły kary:

- małe `C` → silniejsza regularyzacja,
- duże `C` → słabsza regularyzacja.
