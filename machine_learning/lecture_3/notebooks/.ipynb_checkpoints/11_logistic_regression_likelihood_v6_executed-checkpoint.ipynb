# 11. Regresja logistyczna: logit, sigmoid, likelihood i interpretacja współczynników

Ten notebook buduje regresję logistyczną od zera. Najważniejsza intuicja:

> Regresja logistyczna nie modeluje bezpośrednio $p$. Modeluje **log-odds**, a potem zamienia je na prawdopodobieństwo przez sigmoid.

Model:

$$
z_i = x_i^T\beta
$$

$$
p_i = P(y_i=1|x_i)=\sigma(z_i)=\frac{1}{1+e^{-z_i}}
$$

$$
logit(p_i)=\log\left(\frac{p_i}{1-p_i}\right)=x_i^T\beta
$$

Dopasowanie nie polega na minimalizacji $SSE$, tylko na maksymalizacji likelihoodu Bernoulliego.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2, norm

pd.set_option('display.precision', 6)
```

## 1. Dane: prosty przykład dydaktyczny

Wyobraźmy sobie, że mierzymy poziom ekspresji genu i chcemy przewidzieć, czy u myszy wystąpił fenotyp chorobowy.

Dane są małe i trochę „nieidealne”, żeby nie było perfekcyjnej separacji. Dzięki temu maksimum likelihoodu jest skończone.

```python
df = pd.DataFrame({
    'gene_expression': [0.5, 0.8, 1.0, 1.3, 1.6, 1.9, 2.1, 2.4, 2.7, 3.0, 3.3, 3.6, 3.9, 4.2],
    'disease':         [0,   0,   0,   0,   1,   0,   1,   0,   1,   1,   0,   1,   1,   1]
})
df
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(df['gene_expression'], df['disease'])
ax.set_xlabel('gene expression')
ax.set_ylabel('disease')
ax.set_yticks([0, 1])
ax.set_title('Dane binarne: liniowa predykcja prawdopodobieństwa nie wystarcza')
ax.grid(True, alpha=0.3)
plt.show()
```

## 2. Funkcje: sigmoid, log-likelihood i IRLS/Newton

Dla pojedynczej obserwacji Bernoulliego:

$$
P(y_i|p_i)=p_i^{y_i}(1-p_i)^{1-y_i}
$$

Likelihood dla całego zbioru:

$$
L(\beta)=\prod_i p_i^{y_i}(1-p_i)^{1-y_i}
$$

Log-likelihood:

$$
\ell(\beta)=\sum_i \left[y_i\log(p_i)+(1-y_i)\log(1-p_i)\right]
$$

W stabilnej numerycznie postaci, jeśli $\eta_i=x_i^T\beta$:

$$
\ell(\beta)=\sum_i \left[y_i\eta_i - \log(1+e^{\eta_i})\right]
$$

Minimalizowana funkcja kosztu to często ujemny log-likelihood:

$$
NLL(\beta)=-\ell(\beta)
$$

To jest ta sama idea co binary cross-entropy.

```python
def sigmoid(z):
    z = np.asarray(z, dtype=float)
    z = np.clip(z, -35, 35)
    return 1 / (1 + np.exp(-z))


def loglik_logistic(beta, X, y):
    eta = X @ beta
    return float(np.sum(y * eta - np.logaddexp(0, eta)))


def fit_logistic_irls(X, y, l2=0.0, max_iter=100, tol=1e-10):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)
    n, k = X.shape
    beta = np.zeros(k)
    penalty = np.eye(k)
    penalty[0, 0] = 0.0
    converged = False
    for iteration in range(max_iter):
        eta = X @ beta
        p = sigmoid(eta)
        W = p * (1 - p)
        gradient = X.T @ (y - p) - l2 * (penalty @ beta)
        information = (X.T * W) @ X + l2 * penalty
        step = np.linalg.pinv(information) @ gradient
        beta_new = beta + step
        if np.max(np.abs(step)) < tol:
            beta = beta_new
            converged = True
            break
        beta = beta_new
    eta = X @ beta
    p = sigmoid(eta)
    W = p * (1 - p)
    information = (X.T * W) @ X + l2 * penalty
    cov_beta = np.linalg.pinv(information)
    se_beta = np.sqrt(np.diag(cov_beta))
    return {
        'beta': beta,
        'se': se_beta,
        'cov': cov_beta,
        'll': loglik_logistic(beta, X, y),
        'p_hat': p,
        'converged': converged,
        'iterations': iteration + 1,
    }
```

## 3. Dopasowanie modelu

Design matrix z interceptem:

$$
X =
\begin{bmatrix}
1 & expression_1 \\
1 & expression_2 \\
\vdots & \vdots
\end{bmatrix}
$$

Model:

$$
logit(P(disease=1))=\beta_0+\beta_1\cdot expression
$$

```python
X = np.column_stack([np.ones(len(df)), df['gene_expression'].to_numpy()])
y = df['disease'].to_numpy()
fit = fit_logistic_irls(X, y)
fit['converged'], fit['iterations']
```

```python
coef_table = pd.DataFrame({
    'coef': fit['beta'],
    'std_error': fit['se'],
}, index=['intercept', 'gene_expression'])
coef_table['z_Wald'] = coef_table['coef'] / coef_table['std_error']
coef_table['p_Wald'] = 2 * norm.sf(np.abs(coef_table['z_Wald']))
coef_table['exp(coef)'] = np.exp(coef_table['coef'])
coef_table
```

Interpretacja współczynnika przy `gene_expression`:

$$
\exp(\beta_1)
$$

jest mnożnikiem odds przy wzroście ekspresji genu o 1 jednostkę. Jeśli $\exp(\beta_1)=2$, to odds rosną dwukrotnie. Jeśli $\exp(\beta_1)=0.5$, to odds maleją o połowę.

```python
df_pred = df.copy()
df_pred['logit_hat'] = X @ fit['beta']
df_pred['p_hat'] = fit['p_hat']
df_pred
```

```python
x_grid = np.linspace(df['gene_expression'].min() - 0.2, df['gene_expression'].max() + 0.2, 300)
X_grid = np.column_stack([np.ones_like(x_grid), x_grid])
p_grid = sigmoid(X_grid @ fit['beta'])

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(df['gene_expression'], df['disease'], label='obserwacje')
ax.plot(x_grid, p_grid, label='dopasowane P(disease=1)')
ax.set_xlabel('gene expression')
ax.set_ylabel('probability / class')
ax.set_title('Regresja logistyczna: sigmoid na osi prawdopodobieństwa')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## 4. Likelihood: porównanie kandydatów

Na małej tabeli pokażemy, że różne parametry dają różne log-likelihoody. Im większy log-likelihood, tym lepiej model przypisuje wysokie prawdopodobieństwa obserwowanym klasom.

Pamiętaj:

$$
NLL=-\ell
$$

czyli maksymalizacja log-likelihoodu jest tym samym co minimalizacja ujemnego log-likelihoodu.

```python
candidates = pd.DataFrame({
    'beta_0': [-3.0, -2.0, fit['beta'][0], -1.0, 0.0],
    'beta_1': [ 0.7,  1.0, fit['beta'][1],  1.2, 0.0],
})
candidates['log_likelihood'] = [loglik_logistic(row[['beta_0','beta_1']].to_numpy(), X, y) for _, row in candidates.iterrows()]
candidates['NLL'] = -candidates['log_likelihood']
candidates.sort_values('NLL')
```

## 5. Model zerowy, likelihood-ratio test i pseudo-$R^2$

Model zerowy ma tylko intercept, czyli przewiduje wszystkim to samo prawdopodobieństwo:

$$
P(y=1)=\bar y
$$

Porównanie log-likelihoodów:

$$
G^2 = 2(\ell_{full}-\ell_{null})
$$

Przy standardowych założeniach $G^2$ porównujemy z rozkładem $\chi^2$ o liczbie stopni swobody równej różnicy liczby parametrów.

Pseudo-$R^2$ McFaddena:

$$
R^2_{McFadden}=1-\frac{\ell_{full}}{\ell_{null}}
$$

Nie interpretujemy go tak samo jak klasycznego $R^2$ z regresji liniowej. To raczej miara poprawy log-likelihoodu względem modelu zerowego.

```python
y_mean = y.mean()
beta0_null = np.array([np.log(y_mean / (1 - y_mean))])
X_null = np.ones((len(y), 1))
ll_null = loglik_logistic(beta0_null, X_null, y)
ll_full = fit['ll']
G2 = 2 * (ll_full - ll_null)
p_lrt = chi2.sf(G2, df=1)
mcFadden_R2 = 1 - ll_full / ll_null

pd.Series({
    'y_mean': y_mean,
    'log_likelihood_null': ll_null,
    'log_likelihood_full': ll_full,
    'G^2 = 2*(ll_full - ll_null)': G2,
    'df': 1,
    'p_value_LRT': p_lrt,
    'McFadden_pseudo_R2': mcFadden_R2,
})
```

## 6. Klasyfikacja przez próg

Regresja logistyczna daje prawdopodobieństwa. Żeby dostać klasy, wybieramy próg, np. $0.5$:

$$
\hat y = I(\hat p \ge 0.5)
$$

To ważne: **model probabilistyczny i decyzja progowa to dwie różne rzeczy**.

```python
def confusion_counts(y_true, p_hat, threshold=0.5):
    pred = (p_hat >= threshold).astype(int)
    tp = int(((pred == 1) & (y_true == 1)).sum())
    fp = int(((pred == 1) & (y_true == 0)).sum())
    tn = int(((pred == 0) & (y_true == 0)).sum())
    fn = int(((pred == 0) & (y_true == 1)).sum())
    return pd.Series({'TP': tp, 'FP': fp, 'TN': tn, 'FN': fn})

confusion_counts(y, fit['p_hat'], threshold=0.5)
```

## 7. Mini-ćwiczenia

### Ćwiczenie A

Jeśli $z=-2$, to:

$$
p=\frac{1}{1+e^{-(-2)}}=\frac{1}{1+e^2}\approx 0.119
$$

### Ćwiczenie B

Jeśli $\beta_1=0.7$, to wzrost predyktora o 1 jednostkę mnoży odds przez:

$$
e^{0.7}\approx 2.01
$$

### Ćwiczenie C

Dla jednej obserwacji z $y=1$ i $p=0.8$ wkład do log-likelihoodu wynosi:

$$
\log(0.8)\approx -0.223
$$

Dla $y=0$ i $p=0.8$ wkład wynosi:

$$
\log(1-0.8)=\log(0.2)\approx -1.609
$$

Model jest karany, gdy daje wysokie prawdopodobieństwo klasie, która się nie wydarzyła.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Przykładowe dane 1D
X = np.array([[2], [4], [5], [6], [8], [9], [10], [12]])
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Model logistyczny
clf = LogisticRegression()
clf.fit(X, y)

# Siatka x do rysowania krzywej
x_grid = np.linspace(X.min() - 1, X.max() + 1, 300).reshape(-1, 1)
p_grid = clf.predict_proba(x_grid)[:, 1]

# Wykres
plt.figure(figsize=(6, 4))

# Punkty danych
plt.scatter(X[y == 0], y[y == 0], marker='x', s=80, label='klasa 0')
plt.scatter(X[y == 1], y[y == 1], marker='o', s=80, label='klasa 1')

# Krzywa logistyczna
plt.plot(x_grid, p_grid, label='sigmoida logistyczna')

# Threshold 0.5
plt.axhline(0.5, linestyle='--', alpha=0.7, label='threshold p=0.5')

plt.xlabel('cecha x')
plt.ylabel('P(y=1 | x)')
plt.title('Regresja logistyczna: sigmoida + punkty 0/1')
plt.ylim(-0.05, 1.05)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
