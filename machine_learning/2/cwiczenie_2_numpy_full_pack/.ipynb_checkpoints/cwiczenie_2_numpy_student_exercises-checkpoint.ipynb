### Porównanie: pętla w Pythonie vs wektoryzacja (ufuncy NumPy)

Poniżej porównujemy dwie implementacje tej samej funkcji (stabilnej sigmoidy) na tym samym wejściu `x_big`:

- **`sigmoid_loop(x)`**: jawna pętla w Pythonie (każdy element liczony osobno → duży narzut interpretera).
- **`sigmoid_stable(x)`**: wersja wektorowa oparta o **ufuncy NumPy** (np. `np.exp` + działania arytmetyczne na tablicach) oraz maski boolean — pętla wykonywana jest wewnątrz skompilowanego kodu C.

Dla małych tablic różnice mogą być mało widoczne w wartościach bezwzględnych, ale dla dużych danych przewaga ufunców rośnie.

````python
np.random.seed(0)

# 1) wersja pętlowa (baseline do porównania)
def sigmoid_loop(x):
    x = np.asarray(x, dtype=np.float64)
    out = np.empty_like(x, dtype=np.float64)
    for i in range(x.size):
        xi = x[i]
        if xi >= 0:
            out[i] = 1.0 / (1.0 + np.exp(-xi))
        else:
            ex = np.exp(xi)
            out[i] = ex / (1.0 + ex)
    return out

# 2) wersja ufunc / wektorowa: sigmoid_stable(x) powinna być już zdefiniowana powyżej

# 3) duże dane: tu ujawnia się przewaga ufunców
x_big = np.random.randn(1_000_000)

# 4) poprawność: oba podejścia powinny dać ten sam wynik
a = sigmoid_loop(x_big)
b = sigmoid_stable(x_big)
print("same?", np.allclose(a, b, rtol=1e-12, atol=1e-12))

# 5) czas (Jupyter/IPython)
%timeit sigmoid_loop(x_big)
%timeit sigmoid_stable(x_big)
````

### Benchmark softplus (opcjonalnie)

Analogicznie możemy porównać wersję pętlową z wersją wektorową dla softplus.

````python
import numpy as np
np.random.seed(0)

def softplus_loop(x):
    x = np.asarray(x, dtype=np.float64)
    out = np.empty_like(x, dtype=np.float64)
    for i in range(x.size):
        xi = x[i]
        # stabilna wersja w pętli (żeby porównanie było uczciwe numerycznie)
        out[i] = np.log1p(np.exp(-abs(xi))) + max(xi, 0.0)
    return out

x_big = np.random.randn(1_000_000)

# poprawność
a = softplus_loop(x_big)
b = softplus_stable(x_big)
print("same?", np.allclose(a, b, rtol=1e-12, atol=1e-12))

# czas (Jupyter/IPython)
%timeit softplus_loop(x_big)
%timeit softplus_stable(x_big)
````

### “Złe” vs “dobre” użycie ufunców

Ufuncy (np. `np.exp`, `np.abs`, `np.log1p`) dają przewagę, gdy działają na **całej tablicy naraz** (pętla wykonuje się w C).  
Jeśli jednak wywołujemy ufunc **wewnątrz pętli w Pythonie** dla każdego elementu, tracimy korzyść wydajnościową — dominuje narzut interpretera.

````python
np.random.seed(0)
x = np.random.randn(1_000_000)

# ZŁE: pętla w Pythonie + ufunc na skalarkach (narzut Pythona dominuje)
def softplus_bad(x):
    x = np.asarray(x, dtype=np.float64)
    out = np.empty_like(x)
    for i in range(x.size):
        xi = x[i]
        out[i] = np.log1p(np.exp(-abs(xi))) + max(xi, 0.0)  # ufuncy, ale wołane 1e6 razy
    return out

# DOBRE: ufuncy na całej tablicy (bez pętli w Pythonie)
def softplus_good(x):
    x = np.asarray(x, dtype=np.float64)
    return np.log1p(np.exp(-np.abs(x))) + np.maximum(x, 0.0)

# poprawność
a = softplus_bad(x)
b = softplus_good(x)
print("same?", np.allclose(a, b, rtol=1e-12, atol=1e-12))

# czas
%timeit softplus_bad(x)
%timeit softplus_good(x)
````

### Dla chętnych - proszę sprawdzić skalowanie prędkości wykonania obu typów algorytmów zależności od wielkości tablicy.

### Dla chętnych — skalowanie czasu względem rozmiaru tablicy

Sprawdź, jak zmienia się czas wykonania dla wersji:
- pętlowej (Python loop),
- wektorowej (ufuncy NumPy),

gdy zwiększamy rozmiar tablicy `N`. Zapisz wnioski: od jakiego `N` przewaga ufunców jest wyraźna i dlaczego.

````python
def time_one(fn, x, repeat=3):
    # proste "mini-timeit" bez dodatkowych narzutów raportowania
    best = float("inf")
    for _ in range(repeat):
        t0 = time.perf_counter()
        fn(x)
        best = min(best, time.perf_counter() - t0)
    return best

Ns = [] # zdefiniuj wielkość tablic np. [10, 100, 1_000, 10_000, 100_000, 1_000_000]
print(f"{'N':>10s} | {'loop [s]':>10s} | {'ufunc [s]':>10s} | {'speedup':>8s}")
print("-"*50)

for N in Ns:
    x = np.random.randn(N)

    t_loop = time_one(softplus_bad, x, repeat=3) if N <= 100_000 else None  # pętla może być bardzo wolna
    t_vec  = time_one(softplus_good, x, repeat=3)

    if t_loop is None:
        print(f"{N:10d} | {'(skip)':>10s} | {t_vec:10.4g} | {'-':>8s}")
    else:
        print(f"{N:10d} | {t_loop:10.4g} | {t_vec:10.4g} | {t_loop/t_vec:8.1f}x")
````

#### Widzimy, że dla małych tablic narzut wywołań dominuje, ale wraz ze wzrostem rozmiaru danych wersja pętlowa rośnie liniowo z dużą stałą, podczas gdy wersja oparta o ufuncy ma znacznie mniejszy koszt jednostkowy i daje ~90× przyspieszenie.

### Mini-demo: `meshgrid` + ufuncy (bez pętli)

`np.meshgrid` pozwala zamienić dwa wektory współrzędnych (x i y) w dwie macierze `X` i `Y`,
które opisują siatkę punktów w 2D. Potem możemy liczyć funkcje na całej siatce ufuncami NumPy
(np. `np.sin`, `np.exp`, `np.sqrt`) bez żadnych pętli w Pythonie.

````python
import numpy as np

# 1) wektory osi
x = np.linspace(-3, 3, 3)
y = np.linspace(-2, 2, 5)

# 2) siatka 2D (domyślnie indexing='xy')
X, Y = np.meshgrid(x, y)

print("x shape:", x.shape, "\ny shape:", y.shape)
print("x (full):", x, "\ny (full):", y)
print("X shape:", X.shape, "\nY shape:", Y.shape)
print("X (full):", X, "\nY (full):", Y)
# 3) ufuncy na całej siatce: przykład funkcji 2D
R = np.sqrt(X**2 + Y**2)          # ufuncy: power, add, sqrt
Z = np.sin(R) / (1 + R)           # ufuncy: sin, add, divide

print("Z shape:", Z.shape)
print("Z (fragment):\n", Z[:2, :3])
print("Z (full):\n", Z)
print(X+Y)
print(X*Y)
````

````python
print("\n--- TABELKA: wartości (X,Y) na siatce ---")
for i, yi in enumerate(y):
    row = [f"({X[i,j]:+4.1f},{Y[i,j]:+4.1f})" for j in range(len(x))]
    print(f"y={yi:+4.1f} :", " ".join(row))
````

````python
np.set_printoptions(precision=3, suppress=True)
print("\nZ (full):\n", Z)
````

````python
print("\nPunkt siatki [i=0,j=0]: (x,y) =", (X[0,0], Y[0,0]))
print("Punkt siatki [i=4,j=2]: (x,y) =", (X[4,2], Y[4,2]))
````
